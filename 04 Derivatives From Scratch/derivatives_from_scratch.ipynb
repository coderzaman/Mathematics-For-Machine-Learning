# **Derivatives in Machine Learning: The Engine of Optimization**


**(মেশিন লার্নিংয়ে ডেরিভেটিভস: অপটিমাইজেশনের মূল চাবিকাঠি)**

#### **1. Instantaneous Change (তাৎক্ষণিক পরিবর্তন)**
*   **English:** A derivative calculates the exact rate of change at a specific, single moment.
    *   *Example:* If you drive from Dhaka to Savar, your average speed might be 40 km/h. But if you look at your speedometer at exactly 10:05 AM, it might say 60 km/h. That 60 km/h is your **instantaneous rate of change**. In ML, a derivative tells us: *"If I change this specific variable (like Weight) right now by a tiny fraction, exactly how much will my output change?"*
*   **বাংলা:** ডেরিভেটিভ মূলত কোনো কিছুর তাৎক্ষণিক পরিবর্তন মাপে। উদাহরণস্বরূপ, ঢাকা থেকে সাভার যাওয়ার সময় আপনার গাড়ির স্পিডোমিটার ঠিক যে মুহূর্তের গতি দেখায়, ডেরিভেটিভও ঠিক সেভাবেই কাজ করে। মেশিন লার্নিংয়ে এটি আমাদের বলে দেয়: "আমি যদি এখন আমার 'Weight' ($w$) একদম সামান্য একটু পরিবর্তন করি, তাহলে ঠিক এই মুহূর্তে আমার আউটপুট কতটুকু বদলাবে?"

#### **2. Loss Function vs. Cost Function (ভুলের হিসাব)**

Screenshot from 2026-05-03 12-59-06.png

*Look at image. The red line is our prediction ($y = wx + b$), and the blue dots are the real data. The dashed pink lines show the distance between our prediction and reality—this is our error.*
*   **Loss Function:** The error calculated for a **single, individual data point** (just one of those dashed pink lines in your image).
*   **Cost Function:** The total or average error across the **entire dataset** (the combination of all those dashed lines). This value tells us how well our model is performing overall.
*   **বাংলা:** ছবিতে লক্ষ্য করুন, লাল লাইনটা হলো মডেলের প্রেডিকশন আর নীল বিন্দুগুলো হলো আসল ডেটা। মাঝখানের ড্যাশ লাইনগুলো হলো আমাদের ভুল (Error)।
    *   **Loss Function:** শুধুমাত্র একটি ডেটা পয়েন্টের জন্য যে ভুলটা হয়েছে (একটি মাত্র ড্যাশ লাইন)।
    *   **Cost Function:** পুরো ডেটাসেটের সবগুলোর ভুলের সামগ্রিক হিসাব বা গড়। এটি নির্দেশ করে আমাদের মডেলটি মোটের ওপর কতটা ভালো বা খারাপ কাজ করছে।

#### **3. Optimization & The Role of Derivatives (অপটিমাইজেশন এবং ডেরিভেটিভের কাজ)**
*   **English:** The ultimate goal of training a model (both ML and DL) is **Optimization**—specifically, minimizing the Cost Function. We want those dashed lines in your image to be as short as mathematically possible.
    *   We use the derivative to check the instantaneous change of the Cost Function relative to the weights. It acts as a compass, telling the optimizer: *"If you increase the weight, the cost will go up. So, decrease the weight to reach the minimum error!"*
*   **বাংলা:** মেশিন লার্নিং এবং ডিপ লার্নিংয়ের মূল লক্ষ্যই হলো Cost Function-কে মিনিমাইজ (সবচেয়ে কম) করা, যাতে আপনার গ্রাফের ড্যাশ লাইনগুলো (Error) একদম ছোট হয়ে যায়।
    *   ডেরিভেটিভ এখানে একটি কম্পাসের মতো কাজ করে। এটি অপটিমাইজারকে বলে দেয় যে, ওয়েট ($w$) বা বায়াস ($b$) বাড়ালে এরর বাড়বে নাকি কমবে। এই তাৎক্ষণিক ঢাল (Slope) বা ডেরিভেটিভ হিসাব করেই মডেল বুঝতে পারে তাকে ওয়েট কোন দিকে পরিবর্তন করতে হবে।


# **How a Model Trains: The Big Picture**

A Machine Learning model (like Linear Regression) essentially learns in 3 main steps:
Screenshot from 2026-05-03 14-26-04.png
1.  **Parameters (The Drivers):** The core drivers of the model are the **Weight ($w$)** and **Bias ($b$)**. In graphical terms, weight is the slope of the line, and bias is where the line crosses the y-axis (intercept). The model's entire job is to find the most "perfect" values for these two parameters.
2.  **Cost Function (Measuring Mistakes):** We start with a random weight and bias. Then, we calculate the **Cost Function (e.g., MSE)** to measure how far our prediction line is from the actual, real-world data points. A higher cost means the model is making a lot of mistakes.
3.  **Derivatives (The Guide):** To reduce those mistakes, the model must change its weight and bias. But should it increase or decrease them? The **Derivative** (or Gradient) makes this decision. By calculating the instantaneous rate of change, the derivative tells the model: *"If you decrease the weight slightly, the error will go down."*

By continuously following the mathematical guidance of the derivative, the model updates its weight and bias step-by-step. When it finally reaches a point where the Cost Function is at its absolute lowest (the Global Minimum), we say the model is **"Trained!"**


### **How a Model Trains: The Big Picture**
**(মডেল কীভাবে শেখে: মূল প্রক্রিয়া)**

মেশিন লার্নিংয়ের লিনিয়ার রিগ্রেশন (Linear Regression) মডেল মূলত ৩টি ধাপে শেখে:

1.  **Parameters (প্যারামিটার):** মডেলের মূল চালিকাশক্তি হলো **Weight ($w$)** এবং **Bias ($b$)**। গ্রাফের ভাষায় ওয়েট হলো লাইনের ঢাল (Slope) এবং বায়াস হলো লাইনটি কত উপরে বা নিচে আছে (Intercept)। মডেলের কাজ হলো এই দুটির সবচেয়ে পারফেক্ট মান খুঁজে বের করা।
2.  **Cost Function (ভুলের পরিমাপ):** আমরা যেকোনো একটি ওয়েট এবং বায়াস দিয়ে শুরু করি। এরপর **Cost Function (যেমন: MSE)** হিসাব করে দেখি, আমাদের প্রেডিকশন লাইনটি আসল ডেটা থেকে কতটা দূরে আছে। কস্ট ফাংশনের মান যত বেশি, মডেলের ভুল তত বেশি।
3.  **Derivatives (পথপ্রদর্শক):** ভুল কমানোর জন্য ওয়েট এবং বায়াস পরিবর্তন করতে হবে। কিন্তু কোন দিকে? বাড়াবো নাকি কমাবো? এই সিদ্ধান্তটি দেয় **Derivative** বা গ্রেডিয়েন্ট (Gradient)। ডেরিভেটিভ তাৎক্ষণিক পরিবর্তনের হার মেপে বলে দেয় যে, ওয়েট একটু কমালে বা বায়াস একটু বাড়ালে এরর (Cost) কমবে।

এইভাবে ডেরিভেটিভের নির্দেশনা মেনে ওয়েট এবং বায়াস আপডেট করতে করতে মডেল যখন এমন এক জায়গায় পৌঁছায় যেখানে Cost Function-এর মান সবচেয়ে কম (Global Minimum), তখন আমরা বলি **"মডেলটি ট্রেইনড (Trained) হয়ে গেছে!"**



# **From Average to Instantaneous Velocity: The Birth of Derivatives**

Screenshot from 2026-05-03 15-51-35.png
**(গড় বেগ থেকে তাৎক্ষণিক বেগ: ডেরিভেটিভের ধারণা)**

When a vehicle starts its journey, its initial velocity is zero ($v = 0$). When it reaches its destination and stops, its final velocity is also zero ($v = 0$). However, the velocity is constantly changing during the journey—speeding up, slowing down, or maintaining a steady pace.
*(বাংলা: একটি গাড়ি যখন যাত্রা শুরু করে তখন তার বেগ শূন্য থাকে, আবার গন্তব্যে পৌঁছালে বেগ শূন্য হয়ে যায়। কিন্তু মাঝখানের পুরোটা সময় এর বেগ অনবরত বদলাতে থাকে।)*

To find the exact speed of the car at a specific millisecond, we use **Derivatives**. Here is how we get there step-by-step:

#### **Step 1: The Large Interval (Average Velocity / গড় বেগ)**
Let's look at the position ($m$) of the car recorded at $5$-second intervals ($t$).

| Time / সময় ($t$ in seconds) | Distance / দূরত্ব ($m$ in meters) |
| :--- | :--- |
| 5 | 10 |
| **10** | **18** |
| **15** | **28** |
| 20 | 32 |
| 25 | 33 |
| 30 | 40 |

If we want to find the velocity around $12.5$ seconds, the simplest way is to calculate the average velocity between $10s$ and $15s$.
*   **Formula:** $\text{Average Velocity} = \frac{\text{Change in Distance}}{\text{Change in Time}} = \frac{\Delta d}{\Delta t}$
*   **Calculation:** $\frac{28 - 18}{15 - 10} = \frac{10}{5} = \mathbf{2 \text{ ms}^{-1}}$

**The Problem:** This assumes the car moved at exactly $2 \text{ ms}^{-1}$ for that entire $5$ seconds. But in reality, it might have been driving at $1 \text{ ms}^{-1}$ at $11s$ and $3 \text{ ms}^{-1}$ at $14s$. Average velocity hides the truth!
*(বাংলা: সমস্যা হলো, এই ৫ সেকেন্ডের বড় ব্যবধানে গড় বেগ বের করলে আমরা আসল চিত্রটি পাই না। হতে পারে ১০ থেকে ১১ সেকেন্ডের মাঝে গাড়ির বেগ এক রকম ছিল, আবার ১৩ থেকে ১৪ সেকেন্ডের মাঝে আরেক রকম ছিল।)*

#### **Step 2: Shrinking the Interval (ছোট ব্যবধান)**
To get a more accurate picture, we record the data every $1$ second instead of every $5$ seconds.

| Time / সময় ($t$ in seconds) | Distance / দূরত্ব ($m$ in meters) |
| :--- | :--- |
| 10 | 18 |
| 11 | 20 |
| **12** | **23** |
| **13** | **27** |
| 14 | 28 |

Now, let's calculate the average velocity between $12s$ and $13s$ to get closer to our target of $12.5s$:
*   **Calculation:** $\frac{27 - 23}{13 - 12} = \frac{4}{1} = \mathbf{4 \text{ ms}^{-1}}$

Notice how drastically the velocity changed from $2 \text{ ms}^{-1}$ to $4 \text{ ms}^{-1}$! By making the time interval smaller, we are getting closer to the true speed of the car at that exact moment.

#### **Step 3: The Infinitely Small Interval (The Derivative / তাৎক্ষণিক বেগ)**
If we want the *exact* velocity at exactly $t = 12.5s$, a $1$-second interval is still too big. We need to shrink the time interval until it is almost zero (a tiny, tiny portion of time).

| Time / সময় ($t$ in seconds) | Distance / দূরত্ব ($m$ in meters) |
| :--- | :--- |
| 12.499 | ... |
| **12.500** | **Exact Position** |
| 12.501 | ... |

When the change in time ($\Delta t$) becomes infinitesimally small, we stop calling it $\Delta t$ and start calling it $dt$ (which stands for *derivative of time*). The change in distance becomes $dd$.

*   **Instantaneous Velocity (তাৎক্ষণিক বেগ):** $\mathbf{v = \frac{dd}{dt}}$

**Conclusion:**
A derivative ($\frac{dd}{dt}$) is simply the average rate of change ($\frac{\Delta d}{\Delta t}$) calculated over an unimaginably tiny interval. In Machine Learning, this exact same logic applies: we calculate how much the Error changes ($\Delta \text{Error}$) for a microscopic change in the Weight ($\Delta \text{Weight}$) to find the derivative: $\frac{\partial E}{\partial w}$.
*(বাংলা: ডেরিভেটিভ আর কিছুই নয়, এটি হলো একদম ক্ষুদ্রাতি ক্ষুদ্র সময়ের ব্যবধানে দূরত্বের পরিবর্তনের হার। মেশিন লার্নিংয়েও আমরা ঠিক এই কাজটিই করি— ওয়েট (Weight) একদম সামান্য পরিবর্তন করলে লস (Loss/Error) কতটুকু পরিবর্তন হয়, তার তাৎক্ষণিক হার বের করাই হলো অপটিমাইজেশন!)*


# **Derivative: The Geometrical Intuition**

**(ডেরিভেটিভের জ্যামিতিক ধারণা)**

জ্যামিতিক ভাষায় ডেরিভেটিভ হলো— **কোনো বক্ররেখার (Curve) একটি নির্দিষ্ট বিন্দুতে স্পর্শকের (Tangent Line) ঢাল (Slope)।**

বিষয়টি কীভাবে কাজ করে, তা আপনার ডেটা এবং ছবির সাহায্যে ধাপে ধাপে বোঝা যাক:
Screenshot from 2026-05-03 16-21-01.png
#### **Step 1: The Secant Line (ছেদক রেখার ঢাল)**
আপনার গ্রাফে দুটি বিন্দু আছে:
*   বিন্দু A: $(5, 5)$
*   বিন্দু B: $(7, 10)$

আপনি যখন এই দুটি বিন্দুর মধ্যে ঢাল বা Average Velocity বের করলেন:
$$\text{Slope} = \frac{\text{Rise}}{\text{Run}} = \frac{10 - 5}{7 - 5} = \frac{5}{2} = 2.5$$

জ্যামিতিতে, দুটি বিন্দুর ভেতর দিয়ে যে সরলরেখা যায় তাকে বলা হয় **Secant Line (ছেদক রেখা)**। আপনার বের করা $2.5$ হলো মূলত এই ছেদক রেখার ঢাল। এটি আপনাকে ৫ সেকেন্ড থেকে ৭ সেকেন্ডের মাঝের **গড় বেগ** দিয়েছে, ঠিক ৫ সেকেন্ডের মুহূর্তের বেগ নয়।

#### **Step 2: Shrinking the Triangle (ত্রিভুজটি ছোট করা)**
image.png

এখন ধরুন, আপনি ঠিক **$t = 5$** সেকেন্ডে আপনার স্পিড কত ছিল তা জানতে চান।
এর জন্য আপনাকে যা করতে হবে: গ্রাফের উপরের বিন্দুটিকে (B) ধীরে ধীরে নিচের বিন্দুর (A) দিকে টেনে আনতে হবে।
*   প্রথমে B বিন্দুটিকে $7$ থেকে $6$ এ আনুন। ত্রিভুজটি ছোট হবে।
*   এরপর $5.5$ এ আনুন। ত্রিভুজটি আরও ছোট হবে।
*   এরপর $5.1$, তারপর $5.01$ এ আনুন।

আপনি যখন বিন্দু B-কে বিন্দু A-এর একদম গায়ে গায়ে লাগিয়ে ফেলবেন, তখন $t$-এর পরিবর্তন (Run বা $dt$) প্রায় শূন্যের কাছাকাছি চলে আসবে।

#### **Step 3: The Tangent Line (স্পর্শক এবং ডেরিভেটিভ)**
যখন বিন্দু দুটি একদম মিলে গিয়ে একটি মাত্র বিন্দুতে (Point A) পরিণত হয়, তখন সেই রেখাটি আর দুটি বিন্দুকে ছেদ করে না, বরং গ্রাফের ঠিক ওই একটি বিন্দুকে আলতো করে ছুঁয়ে যায়।

*   এই ছুঁয়ে যাওয়া রেখাটিকেই বলা হয় **Tangent Line বা স্পর্শক**।
*   আর এই স্পর্শকের যে ঢাল (Slope), তাকেই ম্যাথের ভাষায় বলা হয় **ডেরিভেটিভ ($dm/dt$)**।

**সংক্ষিপ্ত কথা:** আপনি বড় ত্রিভুজ দিয়ে যে Average Slope বা $\frac{\Delta m}{\Delta t}$ বের করেছিলেন, সেই ত্রিভুজটিকে যখন আপনি অসীমভাবে ছোট করে ফেলেন, তখন সেটি ডেরিভেটিভ বা $\frac{dm}{dt}$ হয়ে যায়। ডেরিভেটিভ মানেই হলো গ্রাফের একটি নির্দিষ্ট বিন্দুর স্পর্শকের ঢাল!


# **Introduction to Functions: The Mathematical Machine**



Before exploring different derivative formulas, we must first understand what a function is.

At its core, a **function** acts like a processing machine. It takes a specific input, applies a mathematical rule to process it, and delivers a specific output.

#### **1. How it Works (The Notation)**
Let's define a function that squares whatever number you give it:

image.png

$$f(x) = x^2$$
*(Alternatively, this is often written as $y = x^2$)*

*   **The Input:** You feed a number into the function. Let's give it the number 2.
*   **The Processing:** The function applies its rule (squaring the input).
*   **The Output:** The machine produces the result.
$$f(2) = 4$$

#### **2. Key Terminology**
When working with functions (and later, their derivatives), we use specific names for our inputs and outputs:

*   **Independent Variable ($x$):** This is your **Input**. It is called "independent" because you are free to choose whatever valid value you want to feed into the function. In the image you uploaded, this is the "Domain" (e.g., 1, 2, 3...).
*   **Dependent Variable ($y$ or $f(x)$):** This is your **Output**. It is called "dependent" because its final value relies entirely on whatever input you chose. In the image you uploaded, this is the "Range" (e.g., 1, 4, 9...).

### **The Derivative of a Constant Function**

Screenshot from 2026-05-03 17-48-56.png

A constant function is a function that always produces the exact same output ($y$), no matter what input ($x$) you give it.

Let's take the function from the graph:
$$y = f(x) = 3$$

| Independent Variable ($x$) | Dependent Variable ($y = f(x)$) |
| :--- | :--- |
| 0 | 3 |
| 1 | 3 |
| 2 | 3 |

#### **1. Calculating the Slope (Average Rate of Change)**
To find the derivative, we are looking for the slope, which is calculated as **Rise / Run** (Change in $y$ divided by Change in $x$).
$$\text{Slope} = \frac{\Delta y}{\Delta x} = \frac{y_2 - y_1}{x_2 - x_1}$$

If we calculate the distance between the points $(1, 3)$ and $(2, 3)$:
$$\frac{\Delta y}{\Delta x} = \frac{3 - 3}{2 - 1} = \frac{0}{1} = \mathbf{0}$$

#### **2. Calculating the Instantaneous Rate of Change (The Derivative)**
Even if we shrink the distance to an infinitesimally small interval (finding the instantaneous change), the result is the same. Let's check the slope between $x = 1$ and a point right next to it, like $x = 1.0000000001$:

$$\frac{\Delta y}{\Delta x} = \frac{3 - 3}{1.0000000001 - 1} = \frac{0}{0.0000000001} = \mathbf{0}$$

#### **3. The Constant Rule in Calculus**
Geometrically, as seen in the graph, $f(x) = 3$ is a perfectly horizontal line. A horizontal line has no steepness, meaning its slope is zero everywhere.

Therefore, we get the first official rule of derivatives: **The derivative of any constant $c$ is always $0$.**

**Mathematical Notation:**
$$\frac{d}{dx}(c) = 0$$

**Examples:**
*   $\frac{d}{dx}(3) = 0$
*   $\frac{d}{dx}(100) = 0$
*   $\frac{d}{dx}(-5.5) = 0$

# **The Derivative of a Straight Line (Linear Functions)**



The equation of a straight line is generally written as $y = mx + c$.
To keep things simple, let's assume the y-intercept is zero ($c = 0$), which gives us:
$$y = mx$$

#### **1. Case 1: When the Slope ($m$) is 1**
Screenshot from 2026-05-03 19-00-19.png
Let $y = f(x) = 1x$ (or simply $y = x$).


| x | y = f(x) |
| :--- | :--- |
| 0 | 0 |
| 1 | 1 |
| 2 | 2 |
| 3 | 3 |

*   **Average Rate of Change (1 unit):**
    $$\text{Slope} = \frac{\Delta y}{\Delta x} = \frac{2 - 1}{2 - 1} = \frac{1}{1} = \mathbf{1}$$
*   **Instantaneous Rate of Change:** Even if we take a microscopic step forward from $x=1$ to $x=1.000001$:

    $$\frac{1.000001 - 1}{1.000001 - 1} = \frac{0.000001}{0.000001} = \mathbf{1}$$
*   *Conclusion:* The rate of change is always constant. The derivative is $1$.

#### **2. Case 2: When the Slope ($m$) is 2**
Screenshot from 2026-05-03 19-05-21.png

Let $y = f(x) = 2x$.

| x| y = f(x) |
| :--- | :--- |
| 0 | 0 |
| 1 | 2 |
| 2 | 4 |
| 3 | 6 |

*   **Average Rate of Change:** Let's calculate between $x=2$ and $x=3$.

    $$\text{Slope} = \frac{\Delta y}{\Delta x} = \frac{6 - 4}{3 - 2} = \frac{2}{1} = \mathbf{2}$$

#### **3. The Algebraic Proof (Using Limits)**
We can prove that the derivative of $y = 2x$ will always be $2$ for *any* point using the definition of a derivative.

Let our starting point be $(x, 2x)$.
Let our new point be a tiny step forward: $(x + \Delta x, 2(x + \Delta x))$.

Now, we find the change in $y$ over the change in $x$:
$$\frac{dy}{dx} = \frac{2(x + \Delta x) - 2x}{(x + \Delta x) - x}$$
$$\frac{dy}{dx} = \frac{2x + 2\Delta x - 2x}{\Delta x}$$
$$\frac{dy}{dx} = \frac{2\Delta x}{\Delta x}$$

The $\Delta x$ cancels out on the top and bottom, leaving us with:
$$\frac{dy}{dx} = \mathbf{2}$$

#### **4. The Linear Rule in Calculus**
Based on these calculations, we can establish a universal rule for straight lines:

**The derivative of any linear term $mx$ is simply its coefficient, $m$.**

**Mathematical Notation:**
$$\frac{d}{dx}(mx) = m$$

**Examples:**
*   $\frac{d}{dx}(5x) = 5$
*   $\frac{d}{dx}(-3x) = -3$
*   $\frac{d}{dx}(x) = 1$
