### **Connecting Derivatives to Gradient Descent: Optimization**

আমরা এতক্ষণ ধরে ডেরিভেটিভ (Derivatives) এবং রেট অব চেঞ্জ (Rate of Change) শিখেছি। কিন্তু মেশিন লার্নিংয়ে এর আসল কাজ কী? এর প্রধান কাজই হলো **Optimization (অপটিমাইজেশন)** বা সবচেয়ে সেরা মানটি খুঁজে বের করা।

#### **1. The Geometry of Maxima and Minima (সর্বোচ্চ এবং সর্বনিম্ন বিন্দু)**
গ্রাফের দিকে তাকালে আমরা দেখতে পাই:
*   **Maximum Point (চূড়া):** রেখাটি যখন উপরের দিকে উঠতে উঠতে সর্বোচ্চ বিন্দুতে পৌঁছায়, তখন মুহূর্তের জন্য এর ওঠা বন্ধ হয়ে যায় এবং এরপর নিচে নামতে শুরু করে।
*   **Minimum Point (তলদেশ):** ঠিক একইভাবে, রেখাটি যখন নিচে নামতে নামতে সর্বনিম্ন বিন্দুতে পৌঁছায়, তখন মুহূর্তের জন্য এর নামা বন্ধ হয়ে যায় এবং আবার উপরে উঠতে শুরু করে।

এই সর্বোচ্চ বা সর্বনিম্ন বিন্দুগুলোতে যদি আমরা একটি Tangent Line বা স্পর্শক আঁকি (ছবিতে লাল রেখা), তাহলে সেটি একদম সমতল (Horizontal) হবে।
যেহেতু রেখাটি সমতল, তাই এখানে কোনো পরিবর্তন বা ঢাল নেই। অর্থাৎ:
$$\text{Slope} = \frac{dy}{dx} = \mathbf{0}$$

#### **2. Derivative for Optimization (অপটিমাইজেশনে ডেরিভেটিভের ব্যবহার)**
যদি আমরা বেগের ($v$) গ্রাফ চিন্তা করি, তবে যে বিন্দুতে ডেরিভেটিভ বা $\frac{dv}{dt}$ এর মান শূন্য ($0$) এর কাছাকাছি চলে আসে, সেখানেই আমরা বেগ বা স্পিডের Maximum অথবা Minimum মানটি পাই।

ম্যাথমেটিক্সে এই ডেরিভেটিভের মান শূন্য ($0$) ধরে কোনো কিছুর সর্বোচ্চ বা সর্বনিম্ন মান বের করার যে প্রক্রিয়া, তাকেই বলা হয় **Optimization**।

#### **3. Machine Learning Connection (কস্ট ফাংশন অপটিমাইজেশন)**
মেশিন লার্নিংয়ে আমাদের মূল লক্ষ্য হলো মডেলের Error বা Cost Function ($J$) কমানো। আমরা এমন একটি পয়েন্ট খুঁজতে চাই যেখানে আমাদের Error হবে সর্বনিম্ন (Minimum)।

*   **Cost Function Graph:** আমরা যদি Cost Function এর একটি গ্রাফ আঁকি, তবে সেটি একটি পাহাড়ের খাদের মতো (U-shape) হবে।
*   **The Goal:** আমাদের লক্ষ্য হলো সেই খাদ বা Minimum পয়েন্টটি খুঁজে বের করা, যেখানে Error সবচেয়ে কম।
*   **How Derivative Helps:** আমরা ওয়েট ($w$) এর সাপেক্ষে Cost Function ($J$) এর ডেরিভেটিভ $\frac{dJ}{dw}$ বের করি। যখন এই ডেরিভেটিভের মান $0$ হয়ে যায়, আমরা নিশ্চিত হই যে আমরা পাহাড়ের একদম নিচে (Minimum Error Point) পৌঁছে গেছি। মডেল অপটিমাইজ হয়ে গেছে!

**সংক্ষিপ্ত কথা:** গ্রেডিয়েন্ট ডিসেন্ট (Gradient Descent) অ্যালগরিদম ধাপে ধাপে ডেরিভেটিভ হিসাব করে এবং মডেলকে সেই বিন্দুর দিকে নিয়ে যায় যেখানে $\frac{dy}{dx} = 0$ হয়। এটিই হলো মেশিন লার্নিংয়ের অপটিমাইজেশনের মূল চাবিকাঠি!

ডেরিভেটিভ নিজে আপনাকে সরাসরি মিনিমাম ভ্যালুটা (y-অক্ষের মান) দেয় না, বরং ডেরিভেটিভ আপনাকে **সেই ম্যাজিক পয়েন্ট বা লোকেশনটা ($x$-অক্ষের মান)** খুঁজে বের করতে সাহায্য করে, যেখানে গেলে আপনি মিনিমাম ভ্যালুটি পাবেন।

চলুন আপনার কথাটিকেই আপনার নিজের শেখা $y = x^2$ ফাংশনের উদাহরণ দিয়ে স্টেপ-বাই-স্টেপ মেলাই:

### **How We Find the Minimum Value (The Optimization Process)**

**Step 1: ডেরিভেটিভ বা Slope এর ইকুয়েশন বের করা**
আমাদের ফাংশন হলো $y = x^2$। এর ডেরিভেটিভ করে আমরা পাই:
$$\frac{dy}{dx} = 2x$$
এই $2x$ হলো আমাদের "Slope Generator" বা ঢাল মাপার মেশিন।

**Step 2: ডেরিভেটিভকে $0$ ধরে $x$ এর মান বের করা (খাদের তলদেশ খোঁজা)**
আমরা গ্রাফ থেকে জানি, একদম নিচে বা মিনিমাম পয়েন্টে কোনো পরিবর্তন নেই, অর্থাৎ ঢাল বা Slope হলো $0$।
তাই আমরা আমাদের ডেরিভেটিভ ইকুয়েশনকে $0$ এর সমান ধরে নিই:
$$2x = 0$$
$$x = 0$$
এই $x = 0$ হলো আপনার সেই **"Point"**! ডেরিভেটিভ আপনাকে বলে দিল, "ভাই, তুমি যদি $x$-অক্ষ বরাবর $0$ পজিশনে যাও, তাহলে তুমি সবচেয়ে নিচু জায়গাটা পাবে।"

**Step 3: y-অক্ষের দিকে লাইন টানা (Minimum Value বের করা)**
আপনার কথামতো, এবার আমরা সেই $x = 0$ পয়েন্ট থেকে $y$-অক্ষের দিকে লাইন টানব (অর্থাৎ মূল ফাংশনে $x$-এর মান বসাব)।
আমাদের মূল ফাংশন: $y = x^2$
এখানে $x = 0$ বসালে:
$$y = 0^2 = 0$$

এই **$y = 0$** হলো আমাদের **Minimum Value** বা সবচেয়ে কম Cost/Error!

***

**সংক্ষিপ্ত সারমর্ম:**
১. **Derivative** আপনাকে সমীকরণ দেয়।
২. সেই সমীকরণ $= 0$ ধরলে আপনি **$x$ এর মান (Location)** পান।
৩. সেই $x$ এর মান মূল ইকুয়েশনে বসালে আপনি **$y$ এর মান (Minimum Cost/Error)** পান।

# **Cost Function Minimization Idea (কস্ট ফাংশন এবং অপটিমাইজেশন)**

মেশিন লার্নিং মডেলে **Cost Function** হলো এমন একটি গাণিতিক উপায়, যার মাধ্যমে আমরা বুঝতে পারি আমাদের মডেলটি সামগ্রিকভাবে (Overall) কতটুকু ভুল করছে। আমাদের মূল লক্ষ্য থাকে এই Cost Function-এর মানকে কমিয়ে একদম সর্বনিম্ন পর্যায়ে নিয়ে আসা (Minimization)।

#### **1. Data Flow & The Dataset**
আমাদের মডেল কীভাবে কাজ করে তার ফ্লো-টি হলো:
`Study Hour (Input) -------> Model ----------> Predict Mark (Output)`

নিচের টেবিলে মডেলের অ্যাকচুয়াল ডেটা, প্রেডিকশন এবং এররের পরিমাণ দেখানো হলো:

*নোট: মডেলে কিছু প্রেডিকশন বাস্তবের চেয়ে বেশি হয়েছে (নেগেটিভ এরর), আবার কিছু কম হয়েছে (পজিটিভ এরর)।*

#### **2. The Cost Function: Mean Squared Error (MSE)**
লিনিয়ার রিগ্রেশনের ক্ষেত্রে মডেলের মোট ভুল বা Error বের করার জন্য আমরা যে কস্ট ফাংশনটি ব্যবহার করি, তার নাম হলো **Mean Squared Error (MSE)** বা $J$।

**Formula:**
$$J = \frac{1}{M} \sum (\text{Actual} - \text{Prediction})^2$$
*(যেখানে $M$ হলো মোট স্যাম্পল বা ডেটার সংখ্যা)*

**কেন স্কয়ার (Square) করা হয়?**
টেবিলে আমরা দেখেছি এরর পজিটিভ এবং নেগেটিভ দুই রকমই হতে পারে। স্কয়ার করার ফলে সব নেগেটিভ এরর পজিটিভ হয়ে যায় এবং বড় ভুলগুলোকে মডেল বেশি গুরুত্ব দিয়ে দেখতে পারে।

#### **3. Linear Model vs. Non-Linear Cost Function**
*   **The Model:** ছবিতে আমরা যে লাল রেখাটি ($y = mx + c$) দেখতে পাচ্ছি, সেটি হলো মডেলের প্রেডিকশন লাইন। এটি একটি সরলরেখা (Linear)।
*   **The Cost Function:** কিন্তু আমরা যখন MSE ফর্মুলাটি দেখি, সেখানে একটি স্কয়ার $(...)^2$ আছে। আমরা আগের লেসনে শিখেছি, স্কয়ার থাকলে সেটি আর সরলরেখা থাকে না, সেটি একটি বক্ররেখা (Curve বা U-shape) হয়ে যায়। অর্থাৎ, **Cost Function একটি Non-linear Function!**

#### **4. Optimization using Derivatives (ডেরিভেটিভের আসল ম্যাজিক)**
যেহেতু Cost Function-টি একটি নন-লিনিয়ার কার্ভ তৈরি করে, তাই এই কার্ভের একটি নির্দিষ্ট **Minimum Point (সর্বনিম্ন বিন্দু)** আছে।

*   এই সর্বনিম্ন বিন্দুতেই মডেলের Error সবচেয়ে কম থাকে।
*   এই বিন্দুটি খুঁজে বের করার জন্যই আমরা **Derivative (রেট অব চেঞ্জ)** ব্যবহার করি।
*   আমরা ডেরিভেটিভ বের করে দেখি ঢাল বা Slope কোন দিকে যাচ্ছে, এবং ধাপে ধাপে (Gradient Descent-এর মাধ্যমে) আমরা সেই পয়েন্টের দিকে এগোই যেখানে ডেরিভেটিভের মান শূন্য ($0$) হয়ে যায়।

যখন ডেরিভেটিভ $0$ হয়, আমরা বুঝে যাই যে আমাদের মডেলটি অপটিমাইজ (Optimize) হয়ে গেছে এবং আমরা বেস্ট প্রেডিকশন লাইনটি পেয়ে গেছি!

# **Gradient Descent Algorithm (Visualization)**

New Project (2).png

Gradient Descent হলো একটি **Optimization Algorithm**, যার সাহায্যে আমরা কোনো ফাংশনের মান (Cost বা Error) ধাপে ধাপে কমিয়ে সর্বনিম্ন (Minimum) পর্যায়ে নিয়ে আসতে পারি।

ধরি, আমাদের একটি ফাংশন আছে $y = f(x)$। আমাদের লক্ষ্য (Goal) হলো $x$-এর এমন একটি মান খুঁজে বের করা, যার জন্য $y$-এর মান সবচেয়ে কম বা সর্বনিম্ন হয়।

#### **1. The Core Idea: আমরা স্টেপ (Step) কীভাবে নেব?**
ওভারঅল $y$-এর মান কমানোর জন্য আমাদের $x$-এর মান পরিবর্তন করে সামনে বা পেছনে স্টেপ নিতে হবে। কিন্তু আমরা কোন দিকে যাব? সেটি বোঝার জন্য আমরা নির্দিষ্ট বিন্দুতে একটি **Tangent Line (স্পর্শক)** আঁকি এবং তার **Rate of Change বা ডেরিভেটিভ $\left(\frac{dy}{dx}\right)$** বের করি।

ঢাল বা রেট অব চেঞ্জ আমাদের বলে দেয় যে $x$ বাড়ালে $y$ বাড়বে নাকি কমবে। যেদিকে গেলে $y$ কমে, আমরা ঠিক সেদিকেই স্টেপ নিই।

#### **2. Two Scenarios of Gradient Descent**
গ্রাফের দিকে লক্ষ করলে আমরা দুটি মূল পরিস্থিতি বা Case দেখতে পাই:

*   **Case A (বাম দিকের ঢাল / Negative Slope):**
    *   A বিন্দুতে যদি আমরা ডেরিভেটিভ $\frac{dy}{dx}$ বের করি, দেখব মানটি Negative (নেগেটিভ)।
    *   **অর্থ:** এর মানে হলো, এখানে $x$-এর মান বাড়ালে $y$-এর মান কমতে থাকে।
    *   **অ্যাকশন:** যেহেতু আমাদের লক্ষ্য $y$ কমানো, তাই এই ক্ষেত্রে আমাদের $x$-এর মান **বাড়াতে হবে** (ডান দিকে স্টেপ নিতে হবে)।

*   **Case B (ডান দিকের ঢাল / Positive Slope):**
    *   B বিন্দুতে যদি আমরা ডেরিভেটিভ $\frac{dy}{dx}$ বের করি, দেখব মানটি Positive (পজিটিভ)।
    *   **অর্থ:** এর মানে হলো, এখানে $x$-এর মান বাড়ালে $y$-এর মানও বাড়তে থাকে।
    *   **অ্যাকশন:** যেহেতু আমরা চাই $y$ কমুক, তাই এখানে $x$-এর মান বাড়ানো যাবে না। বুদ্ধিমানের কাজ হলো $x$-এর মান **কমাতে থাকা** (বাম দিকে স্টেপ নেওয়া)।

#### **3. Summary of the Process**
Gradient Descent মূলত $x$-এর একটি র‍্যান্ডম ভ্যালু (Random initial value) নিয়ে যাত্রা শুরু করে। এরপর সে ডেরিভেটিভ চেক করে:
*   Slope নেগেটিভ হলে $\rightarrow$ $x$ বাড়ায়।
*   Slope পজিটিভ হলে $\rightarrow$ $x$ কমায়।

এভাবে ধাপে ধাপে সে এমন একটি বিন্দুতে এসে পৌঁছায় যেখানে ডেরিভেটিভ বা Slope শূন্য ($0$) হয়ে যায়। আর সেটিই হলো আমাদের কাঙ্ক্ষিত **Minimum Point!**


# **Gradient Descent Procedure (গ্রেডিয়েন্ট ডিসেন্টের কার্যপদ্ধতি)**

New Project (2).png

Gradient Descent হলো একটি অপটিমাইজেশন অ্যালগরিদম, যা একটি Iterative Process বা পুনরাবৃত্তিমূলক প্রক্রিয়ার মাধ্যমে কাজ করে। এর মূল লক্ষ্য হলো $x$-এর মান বারবার আপডেট করার মাধ্যমে Cost Function ($y$ বা Error)-এর মান ধাপে ধাপে কমিয়ে সর্বনিম্ন (Minimum) পর্যায়ে নিয়ে আসা।

এখানে মূলত তিনটি Scenario বা অবস্থা তৈরি হতে পারে:

1.  **Scenario 1 (Negative Slope): $x \uparrow \implies y \downarrow$**
    *   **অবস্থা:** এই পয়েন্টে $x$-এর মান বাড়ালে $y$ (Error) কমতে থাকে। (যেমন: গ্রাফের **A** বিন্দু)
    *   **অ্যাকশন:** যেহেতু আমাদের লক্ষ্য $y$ কমানো, তাই আমাদের $x$-এর মান **বাড়াতে হবে**।

2.  **Scenario 2 (Positive Slope): $x \uparrow \implies y \uparrow$**
    *   **অবস্থা:** এই পয়েন্টে $x$-এর মান বাড়ালে $y$ (Error)-ও বাড়তে থাকে। (যেমন: গ্রাফের **B** বিন্দু)
    *   **অ্যাকশন:** যেহেতু আমরা Error বাড়াতে চাই না, তাই এই ক্ষেত্রে আমাদের $x$-এর মান **কমাতে হবে**।

3.  **Scenario 3 (Zero Slope / Minimum): $x$ change $\implies y$ no change**
    *   **অবস্থা:** এখানে $x$-এর মান সামান্য পরিবর্তন করলেও $y$-এর মানের কোনো পরিবর্তন হয় না। অর্থাৎ, ঢাল বা ডেরিভেটিভ শূন্য ($0$)।
    *   **অ্যাকশন:** এর মানে হলো আমরা একদম সর্বনিম্ন বিন্দুতে (Minimum point) পৌঁছে গেছি। আমাদের কাজ এখানে শেষ। এই অবস্থাকে বলা হয় **Convergence (কনভার্জেন্স)**।

### **Mathematical Representation (গাণিতিক সমীকরণ)**

যেহেতু এটি একটি Iterative Process, তাই আমাদের $x$-এর মান বারবার আপডেট করতে হবে।
নতুন $x$ বের করার জন্য আমাদের পুরোনো $x$-এর সাথে কিছু যোগ বা বিয়োগ করতে হবে:
$$x_{\text{new}} = x_{\text{old}} \quad ? \quad \frac{dy}{dx}$$

**লজিক (Inverse Relation):**
*   যখন $\frac{dy}{dx}$ পজিটিভ, তখন $x$ কমাতে হবে (অর্থাৎ, বিয়োগ করতে হবে)।
*   যখন $\frac{dy}{dx}$ নেগেটিভ, তখন $x$ বাড়াতে হবে (অর্থাৎ, যোগ করতে হবে)।

এই উল্টো সম্পর্কটি গাণিতিকভাবে সমাধান করার জন্য আমরা সমীকরণে একটি মাইনাস ($-$) চিহ্ন ব্যবহার করি। তাহলে সমীকরণটি দাঁড়ায়:
$$x := x - \frac{dy}{dx}$$
*(এখানে $:=$ মানে হলো অ্যাসাইন করা বা মান আপডেট করা)*

#### **Mathematical Example (ম্যাথ দিয়ে প্রমাণ)**

ধরি, আমাদের Cost Function হলো $y = x^2$
এর ডেরিভেটিভ: $\frac{dy}{dx} = 2x$

**উদাহরণ ১: (যখন Slope পজিটিভ)**
ধরি, আমরা $x = 3$ বিন্দুতে আছি।
*   ডেরিভেটিভ বা Slope $= 2(3) = \mathbf{6}$ (পজিটিভ, মানে আমরা গ্রাফের ডান দিকে আছি, Error বাড়ছে)।
*   আমাদের সমীকরণ অনুযায়ী আপডেট করি:
    $$x_{\text{new}} = 3 - 6 = \mathbf{-3}$$
*   **ফলাফল:** লক্ষ্য করুন, $x$-এর মান $3$ থেকে কমে $-3$-এর দিকে চলে গেল। অর্থাৎ, সূত্রটি নিজে থেকেই $x$-কে কমিয়ে বাম দিকে (Minimum-এর দিকে) পাঠিয়ে দিল!

**উদাহরণ ২: (যখন Slope নেগেটিভ)**
ধরি, আমরা $x = -2$ বিন্দুতে আছি।
*   ডেরিভেটিভ বা Slope $= 2(-2) = \mathbf{-4}$ (নেগেটিভ, মানে আমরা গ্রাফের বাম দিকে আছি, ডান দিকে গেলে Error কমবে)।
*   আমাদের সমীকরণ অনুযায়ী আপডেট করি:
    $$x_{\text{new}} = -2 - (-4)$$
    $$x_{\text{new}} = -2 + 4 = \mathbf{2}$$
*   **ফলাফল:** এখানে মাইনাসে-মাইনাসে প্লাস হয়ে গেল! $x$-এর মান $-2$ থেকে বেড়ে $2$ হয়ে গেল। অর্থাৎ, সূত্রটি নিজে থেকেই $x$-কে বাড়িয়ে ডান দিকে (Minimum-এর দিকে) নিয়ে এলো!

**সিদ্ধান্ত:** $x := x - \frac{dy}{dx}$ সমীকরণটি সম্পূর্ণ নিখুঁতভাবে কাজ করে এবং এটি সবসময় আমাদের Minimum Point বা কনভার্জেন্সের দিকেই নিয়ে যায়।

# **The Role of Learning Rate ($\alpha$) in Gradient Descent**

New Project (3).png



আমরা আগে দেখেছি Gradient Descent-এর আপডেট রুলটি হলো: $x := x - \frac{dy}{dx}$।
কিন্তু বাস্তবে এই সমীকরণটি সম্পূর্ণ নয়। কারণ, এখানে ডেরিভেটিভের সাথে বাই ডিফল্ট $1$ গুণ অবস্থায় থাকে। এর ফলে, ঢাল বা রেট অব চেঞ্জ যদি অনেক বড় হয়, তবে মডেলটি অনেক বড় একটি স্টেপ নিয়ে ফেলে।

এই স্টেপ বা লাফের আকার নিয়ন্ত্রণ করার জন্য আমরা সমীকরণে নতুন একটি ভেরিয়েবল যুক্ত করি, যাকে বলা হয় **Learning Rate ($\alpha$)**।

**The Complete Update Equation:**
$$x := x - \alpha \left( \frac{dy}{dx} \right)$$
*(এখানে $\alpha$ নির্ধারণ করে দেয় যে, ডেরিভেটিভের মান যাই হোক না কেন, মডেল ঠিক কতটুকু সাইজের স্টেপ নেবে।)*

#### **1. The Problem: Learning Rate না থাকলে কী হতো? (Overshooting)**
ধরি, আমাদের লার্নিং রেট নেই (অর্থাৎ মান $1$)। আমরা গ্রাফের **A** পয়েন্টে আছি।
*   A পয়েন্টে রেট অব চেঞ্জ অনেক খাড়া, তাই ডেরিভেটিভের মান অনেক বড়।
*   সূত্র অনুযায়ী $x$-এর মান অনেকখানি বেড়ে যাবে এবং সেটি Minimum Point পার হয়ে একেবারে ডানদিকের **B** পয়েন্টে চলে যাবে।
*   B পয়েন্টে এসে মডেল দেখবে ঢাল পজিটিভ, তাই সে $x$-এর মান কমানোর জন্য আবার বিশাল এক লাফ দিয়ে বাঁ দিকের **C** পয়েন্টে চলে আসবে।
*   এভাবে মডেলটি Minimum Point-এ না পৌঁছে শুধু এপাশ-ওপাশ (Back and forth) লাফাতে থাকবে। একে বলা হয় **Overshooting**।

#### **2. The Impact of Learning Rate (লার্নিং রেট ছোট বা বড় হলে কী হয়?)**

লার্নিং রেট নির্ধারণ করা খুবই বুদ্ধিমানের কাজ। এটি খুব বড় হলেও সমস্যা, আবার খুব ছোট হলেও সমস্যা।

*   **Case 1: Learning Rate is Too Small (খুব ছোট হলে)**
    *   ধরি, রেট অব চেঞ্জ $= 5$ এবং লার্নিং রেট ($\alpha$) $= 0.00001$
    *   Step Size $= 0.00001 \times 5 = 0.00005$
    *   **সমস্যা:** স্টেপ সাইজ এতই ছোট হবে যে, মডেলটি Minimum Point-এ পৌঁছাতে অনন্তকাল সময় লাগিয়ে দেবে। একে বলা হয় **Slow Convergence**।

*   **Case 2: Learning Rate is Too Large (খুব বড় হলে)**
    *   ধরি, রেট অব চেঞ্জ $= 5$ এবং লার্নিং রেট ($\alpha$) $= 2$
    *   Step Size $= 2 \times 5 = 10$
    *   **সমস্যা:** স্টেপ সাইজ অনেক বড় হয়ে যাবে। ফলে এটি Minimum Point মিস করে এপাশ-ওপাশ লাফাতে থাকবে এবং কখনোই অপটিমাইজ হতে পারবে না। একে বলা হয় **Divergence**।

**সিদ্ধান্ত:**
Gradient Descent-কে সফলভাবে কাজ করানোর জন্য আমাদের এমন একটি **"Just Right"** লার্নিং রেট (যেমন: $0.01$ বা $0.001$) বেছে নিতে হয়, যা খুব ধীরও হবে না, আবার মিনিমাম পয়েন্টকে মিসও করবে না।

# **Use of Gradient Descent in Machine Learning (Linear Regression)**

মেশিন লার্নিংয়ে Gradient Descent-এর আসল কাজ হলো মডেলের ওয়েট (Weights) বা প্যারামিটারগুলো আপডেট করে Error বা Cost Function-কে সর্বনিম্ন পর্যায়ে নিয়ে আসা।

#### **1. The Goal (আমাদের লক্ষ্য)**
আমাদের লক্ষ্য হলো লিনিয়ার রিগ্রেশন ইকুয়েশনের ($y = mx + c$) জন্য এমন একটি "Perfect Line" বা বেস্ট ফিট লাইন তৈরি করা, যেন প্রেডিকশন এবং আসল ডেটার মাঝখানের ভুল (Error) সবচেয়ে কম হয়।

আমাদের মূল Cost Function (MSE) হলো:
$$J = \frac{1}{M} \sum (y - \hat{y})^2$$
আমাদের বের করতে হবে $m$-এর ঠিক কোন মানের জন্য এই Cost Function ($J$)-এর মান সর্বনিম্ন (Minimum) হবে।

#### **3. Updating Weight using Gradient Descent**
ম্যানুয়ালি চেক না করে, অ্যালগরিদম নিজে থেকেই এই মিনিমাম পয়েন্টটি খুঁজে বের করার জন্য Gradient Descent ব্যবহার করে। এর জন্য $m$-এর মান আপডেট করার গাণিতিক সূত্রটি হলো:

$$m := m - \alpha \cdot \frac{d(J)}{dm}$$
*(এখানে $\alpha$ হলো Learning Rate এবং $\frac{d(J)}{dm}$ হলো $m$-এর সাপেক্ষে Cost Function-এর ডেরিভেটিভ বা গ্রেডিয়েন্ট)*

#### **4. How it Reaches the Minimum Point (গ্রাফিক্যাল লজিক)**

আপনার দেওয়া গ্রাফের A এবং B বিন্দুর লজিকটি অ্যালগরিদমে ঠিক কীভাবে কাজ করে:

*   **Point A (বাম দিকের অবস্থা):**
    এখানে আমরা দেখতে পাচ্ছি $m$-এর মান বাড়ালে Cost কমছে (ঢাল বা Slope নেগেটিভ)।
    যেহেতু সূত্রে আগে থেকেই একটি মাইনাস ($-$) আছে, তাই নেগেটিভ ঢালের কারণে মাইনাসে-মাইনাসে প্লাস হয়ে যাবে এবং মডেলটি নিজে থেকেই $m$-এর মান **বাড়িয়ে দেবে** (ডান দিকে যাবে)।

*   **Point B (ডান দিকের অবস্থা):**
    এখানে $m$-এর মান বাড়ালে Cost-ও বাড়ছে (ঢাল বা Slope পজিটিভ)।
    সূত্র অনুযায়ী পজিটিভ ঢাল বসলে মাঝখানের মাইনাসটি মাইনাসই থেকে যাবে, এবং মডেলটি নিজে থেকেই $m$-এর মান **কমিয়ে দেবে** (বাম দিকে যাবে)।

এভাবে ধাপে ধাপে আপডেট হতে হতে, মডেলটি এমন এক জায়গায় এসে থামবে যেখানে Cost Function একদম সর্বনিম্ন (Minimum)। ঠিক সেই পয়েন্টের $m$-এর মানটিই হবে আমাদের লিনিয়ার রিগ্রেশন মডেলের জন্য **Best Weight**!


# **Gradient Descent for Multiple Variables (Multivariate)**

আগে আমরা হিসাব সহজ করার জন্য $c$-কে Constant (ধ্রুবক) ধরে শুধু $m$ (Slope) পরিবর্তন করে Cost Function-এর U-আকৃতির কার্ভ দেখেছিলাম। কিন্তু বাস্তবে, লিনিয়ার রিগ্রেশন মডেলে $m$ এবং $c$ উভয়েই Variable। যখন আমরা এই দুটিকেই একসাথে পরিবর্তন করি, তখন আমাদের Cost Function আর সাধারণ 2D লাইনে থাকে না, সেটি একটি **3D Surface বা বাটি (Bowl)**-এর আকার ধারণ করে।

#### **1. The 3D Cost Function (m এবং c এর গ্রাফ)**
আপনার দেওয়া ছবিতে আমরা দেখতে পাচ্ছি:
New Project (8).png
*   **X-axis (সবুজ):** $m$ বা $\theta_0$ এর মান।
*   **Y-axis (লাল):** $c$ বা $\theta_1$ এর মান।
*   **Z-axis (বেগুনি):** Cost Function বা Error ($J$)-এর মান।

আমাদের লক্ষ্য হলো এই 3D বাটির একদম তলদেশে (Minimum Point-এ) পৌঁছানো, যেখানে Error সবচেয়ে কম।

#### **2. কিভাবে দুইটি ভেরিয়েবল একসাথে চেঞ্জ হবে? (The Core Confusion)**
আপনার প্রশ্ন ছিল: "আমি বুঝিনি কিভাবে দুইটা চেঞ্জ একসাথে হবে।"

**বাস্তব উদাহরণ (The Hill Analogy):**
ধরে নিন আপনি একটি পাহাড়ে দাঁড়িয়ে আছেন (যা 3D)। নিচে নামার জন্য আপনার কাছে দুটি দিক খোলা আছে: সামনে-পেছনে (North-South) এবং ডানে-বামে (East-West)।
*   **North-South** দিকটি হলো আপনার $m$-এর ঢাল (Partial Derivative $\frac{\partial J}{\partial m}$)।
*   **East-West** দিকটি হলো আপনার $c$-এর ঢাল (Partial Derivative $\frac{\partial J}{\partial c}$)।

আপনি যখন পা বাড়াবেন, তখন আপনি শুধু সোজা সামনে বা শুধু সোজা ডানে যাবেন না। আপনি একইসাথে দেখবেন সামনে কতটা ঢাল এবং ডানে কতটা ঢাল। এরপর এই দুই ঢালের ওপর ভিত্তি করে আপনি **কোণাকুণি (Diagonally)** একটি স্টেপ নেবেন।

অ্যালগরিদমও ঠিক এই কাজটিই করে। সে প্রতি স্টেপে $m$-এর রেট অব চেঞ্জ এবং $c$-এর রেট অব চেঞ্জ আলাদা আলাদাভাবে বের করে, তারপর **একই সময়ে (Simultaneously)** দুটির মানই আপডেট করে দেয়!

**Mathematical Representation (Simultaneous Update):**
$$m_{\text{new}} = m_{\text{old}} - \alpha \cdot \frac{\partial J}{\partial m}$$
$$c_{\text{new}} = c_{\text{old}} - \alpha \cdot \frac{\partial J}{\partial c}$$
*(এখানে দুটি সূত্রই একসাথে রান করে এবং মডেল কোণাকুণি বাটির নিচে নামতে থাকে।)*

#### **3. Scaling to Multiple Variables (Multivariate Regression)**
যখন আমাদের ডেটাসেটে ফিচার (Variables) অনেক বেড়ে যায়, তখন মডেলটি কীভাবে কাজ করে তা আপনি একদম ঠিকঠাক ধরেছেন:

*   **2 Variables ($w_1, w_2$ and $b$):**
    সমীকরণ: $y = w_1x_1 + w_2x_2 + b$
    এখানে আমাদের ৩টি জিনিস আপডেট করতে হবে ($w_1, w_2, b$)। যেহেতু ইনপুট ভেরিয়েবল দুটি ($x_1, x_2$), তাই মডেলের প্রেডিকশন লাইনটি আর আগের মতো সরলরেখা (Line) থাকবে না, এটি একটি **Plane (সমতল বা চাদর)** হয়ে যাবে।

*   **3+ Variables ($w_1, w_2, w_3$ and $b$):**
    সমীকরণ: $y = w_1x_1 + w_2x_2 + w_3x_3 + b$
    এখানে ৪টি জিনিস আপডেট করতে হবে ($w_1, w_2, w_3, b$)। যখন ভেরিয়েবল ৩টি বা তার বেশি হয়ে যায়, তখন মানুষের চোখ দিয়ে সেটি আর গ্রাফে আঁকা বা দেখা সম্ভব হয় না। ম্যাথমেটিক্সের ভাষায় 4D বা তার ওপরের ডাইমেনশনের এই মডেলগুলোকে বলা হয় **Hyperplane (হাইপারপ্লেন)**।

**Summary:**
ভেরিয়েবল ১টি হোক বা ১০০০টি—Gradient Descent-এর লজিক সবসময় একই থাকে। সে প্রতিটি ওয়েট ($w_1, w_2, \dots, w_n$)-এর Partial Derivative বের করে এবং একইসাথে সবগুলো ওয়েটকে আপডেট করে Cost Function-কে নিচে (Minimum Point-এ) নামিয়ে আনে।
