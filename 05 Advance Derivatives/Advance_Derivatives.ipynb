# **Derivative of a Non-Linear Function (The Squared Function)**

Screenshot from 2026-05-04 15-27-01.png

**Equation:** $y = x^2$

Let's first look at the values of this function:

| $x$ | y = x^2 |
| :--- | :--- |
| -2 | 4 |
| -1 | 1 |
| 0 | 0 |
| 1 | 1 |
| 2 | 4 |

#### **1. Linear vs. Non-Linear (সরলরেখা বনাম বক্ররেখা)**
*   **Linear Equation:** লিনিয়ার ইকুয়েশনের ক্ষেত্রে আমরা একটি সরলরেখা (Straight line) পাই, যেখানে রেট অব চেঞ্জ (Rate of change) বা ঢাল সবসময় স্থির (Constant) থাকে।
*   **Non-Linear Equation:** এখানে আমরা একটি বক্ররেখা (Curve) পাই। এর প্রধান পার্থক্য হলো— বক্ররেখার ক্ষেত্রে রেট অব চেঞ্জ প্রতিটা বিন্দুতে আলাদা হয় (Different rate of change at different times)।
    *   যেমন, $x = -1$ এর জন্য আমরা দেখতে পাচ্ছি y-এর ভ্যালু কমছে (Rate of change is decreasing)।
    *   আবার, $x = 1$ এর জন্য আমরা দেখতে পাচ্ছি y-এর ভ্যালু বাড়ছে (Rate of change is increasing)।

এখান থেকে বোঝা যায়, নন-লিনিয়ার ইকুয়েশনের ক্ষেত্রে $x$ এর আলাদা আলাদা ভ্যালুর জন্য $y$ এর পরিবর্তনের হারও আলাদা হয়।

#### **2. Shrinking the Interval (Numerical View)**
আমরা যখন 1-unit ব্যবধান ধরে হিসাব করি, তখন কার্ভটি আর সরলরেখা থাকে না, তাই ঢালটি একুরেট আসে না। কিন্তু আমরা যদি $\Delta x$ (Interval) কে ছোট করতে থাকি, তখন সেই ছোট্ট অংশটুকু অনেকটা সরলরেখার মতো আচরণ করে এবং এরর (Error) কমে যায়।

$x = 1$ বিন্দু থেকে শুরু করে ব্যবধান ছোট করার চার্ট:

| $\Delta x$ (Run) | $\Delta y$ (Rise) | Slope ($\frac{\Delta y}{\Delta x}$) |
| :--- | :--- | :--- |
| 1 | 3 | 3 / 1 = **3** |
| 0.5 | 1.25 | 1.25 / 0.5 = **2.5** |
| 0.25 | 0.562 | 0.562 / 0.25 = **2.25** |
| 0.125 | 0.265 | 0.265 / 0.125 = **2.125** |
| 0.0001 | 0.0002 | 0.0002 / 0.0001 = **2.0001** (Tends to 2) |

এখান থেকে আমরা বুঝতে পারি যে, $x = 1$ বিন্দুতে ঢাল **2** এর কাছাকাছি চলে আসছে।
সুতরাং আমরা লিখতে পারি: $\frac{d}{dx}(x^2) = 2 \cdot x$

যেহেতু রেট অব চেঞ্জ প্রতিটা পয়েন্টে আলাদা হতে পারে:
*   If $x = 0$, slope is $2(0) = 0$
*   If $x = -2$, slope is $2(-2) = -4$
*   If $x = 1$, slope is $2(1) = 2$

#### **3. Mathematical View of Quadratic Derivative (বীজগণিতীয় প্রমাণ)**
We know the rate of change is:
$$\text{Slope} = \frac{\Delta y}{\Delta x}$$

যেহেতু আমরা একটি নির্দিষ্ট বিন্দুর জন্য রেট অব চেঞ্জ বের করছি, তাই আমরা $\Delta x \to 0$ ধরে নিই (অর্থাৎ $\Delta x$-এর মান 0 এর খুব কাছাকাছি)।

আমাদের ফাংশন: $y = f(x) = x^2$
আমাদের বিন্দুসমূহ: $(x, x^2)$ এবং $(x + \Delta x, (x + \Delta x)^2)$

এখন ডেরিভেটিভের সমীকরণে মান বসালে পাই:
$$\frac{\Delta y}{\Delta x} = \frac{(x + \Delta x)^2 - x^2}{(x + \Delta x) - x}$$


$$= \frac{x^2 + 2x(\Delta x) + (\Delta x)^2 - x^2}{\Delta x}$$


$$= \frac{2x(\Delta x) + (\Delta x)^2}{\Delta x}$$


$$= \frac{\Delta x(2x + \Delta x)}{\Delta x}$$

$$= 2x + \Delta x$$

যেহেতু $\Delta x$ প্রায় শূন্যের কাছাকাছি ($\Delta x \to 0$), তাই:
$$= \mathbf{2x}$$

# **Derivative of a Cubic Function & The Power Rule**

Screenshot from 2026-05-04 16-27-03.png

**Equation:** $f(x) = x^3$

| $x$ | $y = x^3$ |
| :--- | :--- |
| -2 | -8 |
| -1 | -1 |
| 0 | 0 |
| 1 | 1 |
| 2 | 8 |

**Graphical Observation (গ্রাফ থেকে পর্যবেক্ষণ):**
গ্রাফের দিকে লক্ষ করলে দেখা যায়, $x$-এর মান বাড়ার সাথে সাথে সবসময় $y$-এর মান বাড়ছে (Curve-টি সবসময় উপরের দিকে উঠছে)। অর্থাৎ, এই গ্রাফটি কখনো নিচের দিকে নামে না। এর মানে হলো, এই ফাংশনের **Rate of Change (Derivative) সবসময় Positive (অথবা 0) হবে**, কখনো Negative হবে না।

#### **1. Mathematical Proof (First Principles)**
We know the rate of change is:
$$\text{Slope} = \frac{\Delta y}{\Delta x} \quad (\text{where } \Delta x \to 0)$$

আমাদের ফাংশন: $y = x^3$
আমাদের বিন্দুসমূহ: $(x, x^3)$ এবং $(x + \Delta x, (x + \Delta x)^3)$

এখন ডেরিভেটিভের সমীকরণে মান বসালে পাই:
$$\frac{\Delta y}{\Delta x} = \frac{(x + \Delta x)^3 - x^3}{(x + \Delta x) - x}$$

$$(x + \Delta x)^3 \text{ কে ফর্মুলায় ভাঙলে:}$$
$$= \frac{x^3 + 3x^2(\Delta x) + 3x(\Delta x)^2 + (\Delta x)^3 - x^3}{\Delta x}$$

$$x^3 \text{ এবং } -x^3 \text{ কাটাকাটি গেলে:}$$
$$= \frac{3x^2(\Delta x) + 3x(\Delta x)^2 + (\Delta x)^3}{\Delta x}$$

$$\Delta x \text{ কমন নিয়ে কাটাকাটি করলে:}$$
$$= 3x^2 + 3x(\Delta x) + (\Delta x)^2$$

যেহেতু $\Delta x$ প্রায় শূন্যের কাছাকাছি ($\Delta x \to 0$), তাই $\Delta x$ যুক্ত পদগুলো $0$ হয়ে যাবে:
$$\frac{dy}{dx} = \mathbf{3x^2}$$

#### **2. Rate of Change at Specific Points**
এখান থেকে আমরা বুঝতে পারি, প্রতিটি বিন্দুর জন্য Rate of Change আলাদা হবে এবং মানটি সবসময় পজিটিভ হবে (যেহেতু $x$ এর উপর স্কয়ার আছে)।

*   If $x = 1$, Slope $= 3(1)^2 = \mathbf{3}$
*   If $x = 2$, Slope $= 3(2)^2 = \mathbf{12}$
*   If $x = -2$, Slope $= 3(-2)^2 = 3(4) = \mathbf{12}$

#### **3. Generalizing The Power Rule (পাওয়ার রুলের সূত্রপাত)**
উপরের প্রমাণগুলো থেকে আমরা চমৎকার একটি প্যাটার্ন দেখতে পাই:
*   $\frac{d}{dx}(x^2) = 2x$
*   $\frac{d}{dx}(x^3) = 3x^2$
*   $\frac{d}{dx}(x^4) = 4x^3$

**The Universal Power Rule Formula:**
এই প্যাটার্ন থেকে আমরা যেকোনো পাওয়ারের ডেরিভেটিভ বের করার একটি সাধারণ সূত্র লিখতে পারি:
$$\mathbf{\frac{d}{dx}(x^n) = n \cdot x^{n-1}}$$

# **Derivative of Logarithmic Functions**

Screenshot from 2026-05-04 18-11-49.png

লগারিদম (Logarithm) এর ডেরিভেটিভ বোঝার আগে পলিনোমিয়াল এবং লগারিদমের রেট অব চেঞ্জের মূল পার্থক্যটি বোঝা জরুরি।

#### **1. Recap: Rate of Change in Polynomials**
*   $y = x^2$: রেট অব চেঞ্জ কমতেও পারে, বাড়তেও পারে (Can increase and decrease)।
*   $y = x^3$: রেট অব চেঞ্জ সবসময় বাড়ে (Always increases)।
*   $y = x^4$: রেট অব চেঞ্জ কমতেও পারে, বাড়তেও পারে।
এই পলিনোমিয়াল ফাংশনগুলোতে রেট অব চেঞ্জ বা ঢাল (Slope) খুব স্পষ্ট এবং দ্রুত পরিবর্তন হয়।

#### **2. Logarithm Intuition: লগারিদম আসলে কী?**
লগারিদম হলো এক্সপোনেনশিয়াল (পাওয়ার) ফাংশনের ঠিক বিপরীত।
লগ ফাংশন মূলত বের করে দেয়: **"বেস (Base) এর পাওয়ার কত হলে আউটপুট $x$ হবে?"**

আপনার ভাগ করার কনসেপ্টটি দিয়ে দেখলে: একটি সংখ্যাকে তার বেস দিয়ে ঠিক কতবার ভাগ করলে ১ পাওয়া যায়, সেটাই লগ!
*   $\log_2(4) \rightarrow$ $4/2 = 2 \rightarrow 2/2 = 1$ (২ বার ভাগ করা গেছে, তাই মান $\mathbf{2}$)
*   $\log_2(8) \rightarrow$ $8/2 \rightarrow 4/2 \rightarrow 2/2$ (৩ বার, তাই মান $\mathbf{3}$)
*   $\log_2(16) \rightarrow$ মান $\mathbf{4}$

**পর্যবেক্ষণ:**
$y$ এর মান ১ ইউনিট করে বাড়াতে ($2 \to 3 \to 4$), $x$ এর মানকে দ্বিগুণ হারে বাড়াতে হচ্ছে ($4 \to 8 \to 16$)। অর্থাৎ, $x$ অনেক বেশি পরিবর্তন হলেও $y$ এর পরিবর্তন হয় খুবই সামান্য। এটি একটি Non-linear function যার গ্রাফ ধীরে ধীরে সমতল হয়ে যায়।

#### **3. The Derivative of Log (রেট অব চেঞ্জ)**
যেহেতু $x$ বাড়লে $y$ এর পরিবর্তন খুব সূক্ষ্ম (slow) হতে থাকে, তাই এর ডেরিভেটিভ বা রেট অব চেঞ্জ $x$ এর ব্যস্তানুপাতিক (inversely proportional)।

**সাধারণ সূত্র:**
$$\frac{d}{dx}(\log_a x) = \frac{1}{x \cdot \ln(a)}$$

**Natural Log বা $\ln(x)$ এর ক্ষেত্রে:**
মেশিন লার্নিংয়ে আমরা সাধারণত বেস হিসেবে $e$ ($2.718...$) ব্যবহার করি, যাকে $\ln(x)$ বলা হয়। এর ডেরিভেটিভ সবচেয়ে সহজ:
$$\frac{d}{dx}(\ln x) = \frac{1}{x}$$

*   If $x = 2$, rate of change $= \frac{1}{2} = 0.5$
*   If $x = 4$, rate of change $= \frac{1}{4} = 0.25$
*   If $x = 8$, rate of change $= \frac{1}{8} = 0.125$

**সিদ্ধান্ত:** $x$ এর মান যত বাড়ে, রেট অব চেঞ্জ ($1/x$) তত ছোট বা সূক্ষ্ম হতে থাকে।

#### **4. Domain Restriction: $x$ কেন নেগেটিভ হতে পারে না?**
লগারিদমের ইনপুট ($x$) কখনোই নেগেটিভ বা ঋণাত্মক হতে পারে না। যদি নেগেটিভ হয়, তবে ফাংশন এবং এর রেট অব চেঞ্জ Undefined (অসংজ্ঞায়িত) হয়ে যায়।

**গাণিতিক প্রমাণ (Using Base 2):**
ধরি, $y = \log_2(x)$। এর মানে হলো $2^y = x$।
এখন $y$ এর মান যাই হোক না কেন, $x$ কখনো নেগেটিভ হতে পারে না:
*   $y = 1 \rightarrow 2^1 = 2$ (Positive)
*   $y = 0 \rightarrow 2^0 = 1$ (Positive)
*   $y = -1 \rightarrow 2^{-1} = \frac{1}{2} = 0.5$ (Positive)
*   $y = -2 \rightarrow 2^{-2} = \frac{1}{4} = 0.25$ (Positive)

যেহেতু বেস পজিটিভ, এর পাওয়ার আপনি পজিটিভ, নেগেটিভ বা জিরো যাই দেন না কেন, রেজাল্ট সবসময় জিরোর চেয়ে বড় হবে। তাই $\log(\text{negative})$ বা $\log(0)$ এর কোনো অস্তিত্ব নেই!


# **Derivative Rules: Scalar Multiplication and Sum Rule**

যখন আমরা বড় এবং জটিল সমীকরণ (যেমন পলিনোমিয়াল) নিয়ে কাজ করি, তখন পুরো সমীকরণকে একসাথে ডেরিভেটিভ করা কঠিন। এই কাজটিকে সহজ করার জন্য আমরা দুটি মূল নিয়ম ব্যবহার করি:

#### **1. Scalar Multiplication Rule (স্কেলার বা ধ্রুবক গুণের সূত্র)**
**নিয়ম:** কোনো ভেরিয়েবলের সাথে যদি কোনো কনস্ট্যান্ট বা স্কেলার মান গুণ আকারে থাকে, তবে আমরা সেই স্কেলার মানটিকে ডেরিভেটিভ করব না। সেটিকে ডেরিভেটিভের বাইরে রেখে শুধু ভেরিয়েবল অংশটুকুর ডেরিভেটিভ করতে হবে।

*   **Example 1: $y = 2x$**
    $$\frac{dy}{dx} = \frac{d}{dx}(2x)$$
    $$= 2 \cdot \frac{d}{dx}(x)$$
    $$= 2 \cdot 1 = \mathbf{2}$$

*   **Example 2: $y = 4x^2$**
    $$\frac{dy}{dx} = \frac{d}{dx}(4x^2)$$
    $$= 4 \cdot \frac{d}{dx}(x^2)$$
    $$= 4 \cdot (2x) = \mathbf{8x}$$

#### **2. The Sum Rule (যোগের সূত্র)**
**নিয়ম:** যদি কোনো সমীকরণে একাধিক অংশ যোগ (বা বিয়োগ) আকারে থাকে, তবে পুরো সমীকরণটিকে একসাথে ডেরিভেটিভ না করে, যোগের প্রতিটি অংশকে (Term) আলাদা আলাদাভাবে ডেরিভেটিভ করতে হবে।

*   **Example 1: $y = x + x^2$**
    $$\frac{dy}{dx} = \frac{d}{dx}(x) + \frac{d}{dx}(x^2)$$
    $$= 1 + 2x$$
    *(সাজিয়ে লিখলে: $\mathbf{2x + 1}$)*

*   **Example 2: The Equation of a Line ($y = mx + c$)**
    এখানে $m$ হলো একটি স্কেলার ভ্যালু (যা $x$ এর সাথে গুণ আকারে আছে), এবং $c$ হলো একটি স্বাধীন কনস্ট্যান্ট ভ্যালু (যার সাথে কোনো $x$ নেই)। যেহেতু $x$ এর পরিবর্তনের সাথে $c$ এর কোনো পরিবর্তন হয় না, তাই এর রেট অব চেঞ্জ $0$ হয়ে যাবে।
    $$\frac{dy}{dx} = \frac{d}{dx}(mx) + \frac{d}{dx}(c)$$
    $$= m \cdot \frac{d}{dx}(x) + 0$$
    $$= m \cdot 1 + 0 = \mathbf{m}$$
    *(এটি গাণিতিকভাবে প্রমাণ করে যে একটি লিনিয়ার লাইনের রেট অব চেঞ্জ সবসময় তার Slope বা $m$-এর সমান হয়!)*

*   **Example 3: $y = x^2 + \ln(x)$**
    $$\frac{dy}{dx} = \frac{d}{dx}(x^2) + \frac{d}{dx}(\ln x)$$
    $$= \mathbf{2x + \frac{1}{x}}$$


# **Product rule and Chain rule**

### **1. The Product Rule (গুণের সূত্র)**

যখন দুটি আলাদা ফাংশন (উভয়েই $x$-এর ওপর নির্ভরশীল) একে অপরের সাথে গুণ আকারে থাকে, তখন আমরা Product Rule ব্যবহার করি।

**সূত্র:**
$$\frac{d}{dx}(a \cdot b) = a \cdot \frac{d}{dx}(b) + b \cdot \frac{d}{dx}(a)$$

**আপনার উদাহরণ:** $y = x^2 \cdot \ln(x)$
এখানে $a = x^2$ এবং $b = \ln(x)$।
$$\frac{dy}{dx} = x^2 \cdot \frac{d}{dx}(\ln x) + \ln(x) \cdot \frac{d}{dx}(x^2)$$
$$= x^2 \cdot \left(\frac{1}{x}\right) + \ln(x) \cdot (2x)$$
$$= x + 2x \ln(x)$$

### **2. The Chain Rule (চেইন রুল)**

যখন একটি ফাংশনের ভেতরে আরেকটি ফাংশন থাকে (Nested function), তখন আমরা চেইন রুল ব্যবহার করি।
**নিয়ম:** প্রথমে বাইরের ফাংশনের (Outer function) ডেরিভেটিভ করতে হবে, এরপর গুণ চিহ্ন দিয়ে ভেতরের ফাংশনের (Inner function) ডেরিভেটিভ করতে হবে।

**আপনার উদাহরণ:** $y = \ln(x^2)$
*   Outer Function: $\ln(\text{something}) \rightarrow$ ডেরিভেটিভ হলো $\frac{1}{\text{something}}$
*   Inner Function: $x^2 \rightarrow$ ডেরিভেটিভ হলো $2x$

$$\frac{dy}{dx} = \frac{1}{x^2} \cdot \frac{d}{dx}(x^2)$$
$$= \frac{1}{x^2} \cdot 2x = \mathbf{\frac{2}{x}}$$

#### **এটিকে "Chain Rule" বা চেইন রুল কেন বলে?**
চেইন বা শিকলের অনেকগুলো আংটা (Links) থাকে। একটি আংটার সাথে আরেকটি আংটা যুক্ত থাকে। আপনি যখন শিকলের এক মাথা টানেন, তখন প্রথম আংটা দ্বিতীয়টিকে টানে, দ্বিতীয়টি তৃতীয়টিকে টানে—এভাবে কাজ করে।

ম্যাথের ভাষায়, যখন ফাংশনের ভেতরে ফাংশন থাকে (যেমন $y$ নির্ভর করে $u$ এর ওপর, আর $u$ নির্ভর করে $x$ এর ওপর), তখন ডেরিভেটিভগুলো চেইনের মতো গুণ আকারে যুক্ত হয়:
$$\frac{dy}{dx} = \frac{dy}{du} \cdot \frac{du}{dx}$$
এখানে $du$ গুলো কোণাকুণি চেইনের মতো যুক্ত হয়ে পুরো ইকুয়েশনটিকে ব্যালান্স করে।

### **3. Solving: $\frac{d}{dx} \ln(\ln(x^2))$**

এটি একটি দারুণ প্রশ্ন! এখানে ফাংশনের ভেতরে ফাংশন, তার ভেতরে আরেকটি ফাংশন আছে। অর্থাৎ চেইনের মোট ৩টি আংটা (Links) আছে:
1.  **Outer:** $\ln(\dots)$
2.  **Middle:** $\ln(x^2)$
3.  **Inner:** $x^2$

চলুন স্টেপ-বাই-স্টেপ সমাধান করি:

**Step 1: Outer Function-এর ডেরিভেটিভ**
সবচেয়ে বাইরের $\ln$-এর জন্য ডেরিভেটিভ হবে $\frac{1}{\text{ভেতরের পুরো অংশ}}$:
$$= \frac{1}{\ln(x^2)} \cdot \frac{d}{dx}(\ln(x^2))$$

**Step 2: Middle Function-এর ডেরিভেটিভ**
এখন মাঝের অংশটুকুর ($\ln(x^2)$) ডেরিভেটিভ করতে হবে। আমরা জানি $\ln(\text{something})$-এর ডেরিভেটিভ $\frac{1}{\text{something}}$:
$$= \frac{1}{\ln(x^2)} \cdot \frac{1}{x^2} \cdot \frac{d}{dx}(x^2)$$

**Step 3: Inner Function-এর ডেরিভেটিভ**
সবশেষে একদম ভেতরের $x^2$-এর ডেরিভেটিভ করতে হবে, যা হলো $2x$:
$$= \frac{1}{\ln(x^2)} \cdot \frac{1}{x^2} \cdot 2x$$

**Step 4: সমীকরণটি সাজিয়ে লেখা (Simplification)**
এখন উপরের এবং নিচের মানগুলো গুণ করে কাটাকাটি করি:
$$= \frac{2x}{x^2 \ln(x^2)}$$
$$= \mathbf{\frac{2}{x \ln(x^2)}}$$

***

**💡 Pro Tip (একটি শর্টকাট ম্যাজিক!):**
লগারিদমের একটি নিয়ম আছে যে, পাওয়ার সামনে চলে আসে: $\ln(x^2) = 2\ln(x)$।
আপনি যদি সমীকরণটিকে শুরুতেই একটু সহজ করে নেন, তাহলে উত্তরটা আরও সহজে মিলে যায়:
$y = \ln(2\ln x)$
এটিকে ডেরিভেটিভ করলে সরাসরি উত্তর আসবে: **$\frac{1}{x \ln(x)}$**।
(আর আপনি যে উত্তরটি বের করেছেন $\frac{2}{x \ln(x^2)}$, সেটিতে $x^2$ এর পাওয়ার $2$ সামনে আনলে কাটাকাটি হয়ে ঠিক এটিই তৈরি হয়!)
