# **Intro To Statistics (স্ট্যাটিস্টিকস পরিচিতি)**

স্ট্যাটিস্টিকস হলো গণিতের এমন একটি শাখা, যা ডেটা বা উপাত্ত নিয়ে কাজ করে।

**Scattered Data vs Structured Data:**
*   **Scattered Data (বিক্ষিপ্ত ডেটা):** যখন ডেটা এলোমেলোভাবে ছড়ানো থাকে (যেমন: `৪৫, ১২, ৯৯, ২৩`), তখন আমরা সেখান থেকে কোনো অর্থ বা সিদ্ধান্ত (Insight) বের করতে পারি না।
*   **Structured Data (সাজানো ডেটা):** যখন সেই ডেটাকে সারি (Row) এবং কলামে (Column) সাজানো হয় (যেমন: `StdId | Roll | Marks`), তখন সেটি একটি অর্থবহ রূপ পায় এবং আমরা সহজেই ডেটাটি বুঝতে পারি।

**What Statistics Does? (স্ট্যাটিস্টিকস কী করে?)**
একটি বিজনেসের উদাহরণ দিলে, স্ট্যাটিস্টিকস মূলত ৪টি ধাপে কাজ করে:
1.  **Data Collection:** কাস্টমারদের তথ্য সংগ্রহ করা।
2.  **Data Representation/Organization:** ডেটাগুলোকে টেবিল বা চার্টে সাজানো।
3.  **Analyze Data:** ডেটার ওপর গাণিতিক হিসাব-নিকাশ করা।
4.  **Find Insight:** সিদ্ধান্ত বের করা (যেমন: কাস্টমাররা কোন প্রোডাক্ট সবচেয়ে বেশি কিনছে)।

**Types of Statistics:**
1.  **Descriptive Statistics:** যে ডেটাগুলো আমাদের হাতে আছে, সেগুলোকে সামারি (Summary) করে বা চার্টের মাধ্যমে উপস্থাপন করা।
2.  **Inferential Statistics:** একটি ছোট স্যাম্পল (Sample) ডেটা বিশ্লেষণ করে, সম্পূর্ণ পপুলেশন (Population) বা বড় ডেটাসেট সম্পর্কে কোনো উপসংহার (Conclusion) বা প্রেডিকশন বের করা।

### **Uses of Statistics in Machine Learning**

মেশিন লার্নিং মডেল ডেটার ওপর ভিত্তি করে ট্রেইন হয় এবং নতুন (Unseen) ডেটার জন্য প্রেডিকশন করে।
এখানে একটি প্রবাদ খুব প্রচলিত: **"Garbage In, Garbage Out"**।
*   মডেলকে যদি ভালো এবং গোছানো ডেটা দিই, তবে মডেল ভালো প্রেডিকশন করবে।
*   মডেলকে যদি Noisy বা Garbage (খারাপ) ডেটা দিই, তবে মডেল কনফিউজড হয়ে যাবে এবং ভুল ইনসাইট দেবে।

আমাদের ডেটা ভালো নাকি খারাপ, তা বোঝার জন্য আমরা **Descriptive Statistics**-এর অ্যালগরিদম বা পরিমাপগুলোর সাহায্য নিই।

### **Understanding Data Spread (ডেটার বিস্তৃতি বোঝা)**

#### **1. Central Tendency (কেন্দ্রীয় প্রবণতা)**
ডেটার ঠিক মাঝখানের অবস্থা বোঝার জন্য আমরা Mean (গড়), Median (মধ্যমা) এবং Mode (প্রচুরক) ব্যবহার করি।

#### **2. The Problem with Average (গড় কেন সবসময় সঠিক চিত্র দেয় না?)**
ধরি, একটি ক্লাসে ছাত্রদের প্রাপ্ত নম্বরগুলো হলো: `১০, ১২, ২০` এবং `৯৮, ৯৯, ১০০`।
*   আমরা যদি সব নম্বর যোগ করে ছাত্রসংখ্যা দিয়ে ভাগ দিই, তবে **Average (গড়) আসবে প্রায় ৫৬.৫**।
*   **ভুল ইন্টারপ্রিটেশন:** এই গড় দেখে মনে হতে পারে যে ক্লাসের বেশিরভাগ ছাত্রই মাঝারি মানের (৫৬-এর কাছাকাছি নম্বর পেয়েছে)। কিন্তু বাস্তবে কেউ ৫৬ পায়নি! একদল পেয়েছে একদম কম (১০-২০), আরেক দল পেয়েছে একদম বেশি (৯৮-১০০)।
*   অর্থাৎ, ডেটার মধ্যে যখন চরম মান বা **Outliers** (খুব ছোট বা খুব বড় মান) থাকে, তখন Average বা গড় আমাদের সম্পূর্ণ ভুল ধারণা দিতে পারে।

#### **3. Variance (ভেদাঙ্ক)**
Average-এর এই ধোঁকাবাজি ধরার জন্যই Variance-এর জন্ম!
Variance আমাদের বলে দেয় ডেটাগুলো তাদের Average (গড়) থেকে গড়ে কতটুকু দূরে ছড়ানো আছে।
*   সূত্র অনুযায়ী, এটি প্রতিটি পয়েন্ট এবং গড়ের পার্থক্যের বর্গের যোগফলের গড় বের করে।
*   সহজ কথায়, ডেটাগুলো গড়ের কাছাকাছি থাকলে Variance কম হয়, আর ডেটাগুলো গড়ের চেয়ে অনেক দূরে দূরে (যেমন: ১০ আর ১০০) থাকলে Variance অনেক বেশি হয়।

#### **4. Standard Deviation (পরিমিত ব্যবধান)**
Variance-এর একটি সমস্যা হলো, এটি মানগুলোকে স্কয়ার (Square) করে দেয়। তাই আসল ডেটার সাথে হিসাব মেলানোর জন্য আমরা Variance-কে রুট (Square Root) করি। একেই বলে **Standard Deviation (SD)** বা $\sigma$।

**মেশিন লার্নিংয়ে Standard Deviation-এর গুরুত্ব:**
*   **Low Standard Deviation (কম হলে):** এর মানে হলো ডেটাগুলো গড়ের খুব কাছাকাছি অবস্থান করছে। ডেটার মধ্যে কোনো অস্বাভাবিকতা নেই। মেশিন লার্নিং মডেল এ ধরনের গোছানো ডেটা থেকে খুব সহজেই প্যাটার্ন শিখতে পারে (Good Data)।
*   **High Standard Deviation (বেশি হলে):** এর মানে হলো ডেটাগুলো অনেক বেশি ছড়ানো-ছিটানো (যেমন: ১০, ২০ আবার ৯৯, ১০০)। ডেটার মধ্যে অনেক নয়েজ (Noise) বা চরম মান আছে। মডেল এই ডেটা থেকে প্যাটার্ন শিখতে গিয়ে অনেক কনফিউজড হয়ে যায় (Bad Data)।

সংক্ষেপে: **Standard Deviation যত কম, মেশিন লার্নিংয়ের জন্য সেই ডেটা তত ভালো!**

# **Use of Statistics in Data Preprocessing (ডেটা প্রস্তুতিতে স্ট্যাটিস্টিকস)**

মেশিন লার্নিং মডেলকে ট্রেইন করার আগে কাঁচা ডেটাকে (Raw Data) মডেলের ব্যবহারোপযোগী করে তোলার প্রক্রিয়াকে **Data Preprocessing** বলা হয়। এই ধাপে স্ট্যাটিস্টিকস সবচেয়ে বড় ভূমিকা পালন করে। এর প্রধান কাজগুলো হলো:


#### **1. Outlier Detection and Removal (আউটলায়ার শনাক্তকরণ ও বাদ দেওয়া)**

New Project (9).png

*   **Outlier কী?** ডেটাসেটের মধ্যে যে মানগুলো সাধারণ রেঞ্জের চেয়ে অস্বাভাবিকভাবে ছোট বা বড়, তাদের Outlier বলে।
    *   *উদাহরণ:* ধরি, একটি ক্লাসের বেশিরভাগ স্টুডেন্টের মার্কস ৫০-৭০ এর মধ্যে। কিন্তু কয়েকজনের মার্কস ১, ২, ৩ (খুব ছোট) এবং কয়েকজনের ৯৮, ৯৯, ১০০ (খুব বড়)। এই অস্বাভাবিক মানগুলোই হলো আউটলায়ার।
*   **কেন বাদ দেব?** আউটলায়ারসহ মডেল ট্রেইন করালে মডেলটি বায়াসড (Biased) বা বিভ্রান্ত হয়ে যায় এবং সাধারণ ডেটার জন্য ভুল প্রেডিকশন দেয়।
*   **সনাক্ত করার উপায় (Statistical Methods):**
    *   **IQR (Interquartile Range):**  ছবিতে, এই পদ্ধতিতে ডেটার মাঝখানের অংশকে (Q1 থেকে Q3) টার্গেট করা হয়। এরপর এর সাথে একটি নির্দিষ্ট মাত্রা (Whisker) যোগ ও বিয়োগ করে একটি বাউন্ডারি তৈরি করা হয়। বাউন্ডারির বাইরের ডেটাগুলোকে (লাল দাগের অংশ) আউটলায়ার হিসেবে ফেলে দেওয়া হয়।
    *   **Z-Score:** এটি মাপে একটি ডেটা তার গড় (Mean) থেকে কত Standard Deviation দূরে আছে। সাধারণত Z-score ৩-এর বেশি বা -৩-এর কম হলে তাকে আউটলায়ার ধরা হয়।

#### **2. Encoding (ডেটাকে সংখ্যায় রূপান্তর)**
মেশিন লার্নিং মডেল টেক্সট বা ছবি বোঝে না, সে শুধু সংখ্যা (Numbers/Math) বোঝে। তাই স্ট্যাটিস্টিক্যাল এনকোডিংয়ের মাধ্যমে অন্যান্য ডেটাকে নাম্বারে কনভার্ট করা হয়।
*   **Text Data:** বিভিন্ন টেক্সটকে (যেমন: শহরের নাম) ০, ১, ২ ইত্যাদি সংখ্যায় কনভার্ট করা হয় (Label Encoding বা One-Hot Encoding)।
*   **Image Data:** ছবিকে পিক্সেল ভ্যালুতে রূপান্তর করা হয়। সাদাকালো ছবির ক্ষেত্রে ০-১ এবং রঙিন (RGB) ছবির ক্ষেত্রে প্রতিটি কালার চ্যানেলকে ০-২৫৫ এর স্কেলে প্রকাশ করা হয়।

#### **3. Feature Scaling (ফিচারগুলোর মান সমতায় আনা)**
*   **সমস্যা:** ধরি আমাদের ডেটাসেটে দুটি ফিচার আছে:
    *   Age: ৩০ - ৭০
    *   Salary: ২০,০০০ - ৫০,০০০
    এখানে Salary-এর সংখ্যাগুলো Age-এর চেয়ে অনেক বড়। মেশিন লার্নিং মডেল (যেমন লিনিয়ার রিগ্রেশন বা ডিস্ট্যান্স-বেসড অ্যালগরিদম) মনে করতে পারে যে, যেহেতু Salary-এর সংখ্যা বড়, তাই এর গুরুত্ব বা Weight-ও বেশি! ফলে মডেল Age-এর প্রভাবকে ছোট করে দেখবে।
*   **সমাধান (Scaling):** এই বৈষম্য দূর করতে স্ট্যাটিস্টিকসের সাহায্যে সবগুলো ফিচারকে একটি নির্দিষ্ট স্কেলে (যেমন: ০ থেকে ১ এর মধ্যে, বা গড়ের কাছাকাছি) নিয়ে আসা হয়। এতে সবগুলো ফিচার সমান গুরুত্ব পায়। এর জন্য সাধারণত **Standardization (Z-score)** বা **Min-Max Normalization** ব্যবহার করা হয়।

#### **4. Feature Selection (প্রয়োজনীয় ফিচার বেছে নেওয়া)**
মডেলে অতিরিক্ত বা অপ্রয়োজনীয় ডেটা দিলে মডেল স্লো হয়ে যায় এবং ভুল শিখতে পারে। তাই কোন ফিচারটি টার্গেটের সাথে যুক্ত, তা বের করতে আমরা স্ট্যাটিস্টিক্যাল **Correlation (সহসম্পর্ক)** ব্যবহার করি।
*   *উদাহরণ:* Age, Salary, T-shirt Color $\rightarrow$ Target (Spending)
    *   **Positive Correlation (পজিটিভ সম্পর্ক):** Salary বাড়লে যদি Spending বাড়ে, তবে এটি পজিটিভ কোরিলেশন। এই ফিচারটি মডেলের জন্য গুরুত্বপূর্ণ।
    *   **Negative Correlation (নেগেটিভ সম্পর্ক):** Age বাড়লে যদি Spending কমে যায়, তবে এটি নেগেটিভ কোরিলেশন। এটিও মডেলের জন্য খুব গুরুত্বপূর্ণ (বিপরীতমুখী প্যাটার্ন)।
    *   **Zero/No Correlation:** T-shirt Color-এর সাথে একজন মানুষের Spending-এর কোনো সম্পর্ক নেই। তাই Correlation চেক করে আমরা বুঝতে পারি যে এটি একটি অপ্রয়োজনীয় ফিচার, এবং এটি মডেল থেকে বাদ (Drop) দিয়ে দেওয়া উচিত।


# **Inferential Statistics: Population vs. Sample**


Inferential Statistics-এর মূল কাজই হলো একটি ছোট ডেটাসেট (Sample) অ্যানালাইজ করে সম্পূর্ণ ডেটাসেট (Population) সম্পর্কে সিদ্ধান্ত বা উপসংহার (Conclusion) বের করা।

#### **1. Population (সমগ্রক)**
*   **সংজ্ঞা:** আমরা যে বিষয়টি নিয়ে অ্যানালাইসিস করতে চাই, তার আওতাভুক্ত **সবগুলো উপাদানের সম্পূর্ণ সেটকে** Population বলে।
*   **প্রতীক:** এর আকার বা সংখ্যাকে সাধারণত $N$ (বড় হাতের N) দ্বারা প্রকাশ করা হয়।
*   **ML Context:** বাস্তবে সম্পূর্ণ Population ডেটা কালেক্ট করা প্রায় অসম্ভব। (যেমন: পৃথিবীর সমস্ত কাস্টমারের ডেটা বা ভবিষ্যতে হতে পারে এমন সব ই-কমার্স ট্রানজেকশন)।

#### **2. Sample (নমুনা)**
*   **সংজ্ঞা:** সম্পূর্ণ Population থেকে অ্যানালাইসিসের জন্য বেছে নেওয়া একটি **ছোট অংশ বা সাবসেটকে (Subset)** Sample বলে।
*   **প্রতীক:** এর আকারকে সাধারণত $n$ (ছোট হাতের n) দ্বারা প্রকাশ করা হয়।
*   **ML Context:** আমরা মেশিন লার্নিংয়ের জন্য যে ডেটাসেট (Dataset) কালেক্ট করি এবং যেটির ওপর মডেল ট্রেইন করাই, সেটি মূলত একটি Sample ডেটা।

#### **3. The Connection in Machine Learning**
*   মেশিন লার্নিং মডেল **Sample Data** ($n$) থেকে ডেটার প্যাটার্ন শেখে।
*   এরপর সেই শেখা প্যাটার্নের ওপর ভিত্তি করে মডেলটি সম্পূর্ণ **Population** ($N$)-এর (বা আনসিন ডেটার) জন্য প্রেডিকশন বা ইনফারেন্স (Inference) তৈরি করে।
*   **শর্ত:** Sample ডেটাকে অবশ্যই Population-এর সঠিক প্রতিনিধিত্ব (Representation) করতে হবে। Sample যদি বায়াসড হয় (যেমন: শুধু এক বয়সের মানুষের ডেটা), তবে মডেলের প্রেডিকশনও Population-এর জন্য ভুল হবে।


# **Handling Missing Values using Mean, Median, and Mode**

কোন পদ্ধতিতে Null value ফিল করব, তা সম্পূর্ণ নির্ভর করে ডেটার ধরন (Type) এবং ডেটাতে আউটলায়ার (Outlier) আছে কি না, তার ওপর।

#### **1. Mean (গড়) দিয়ে কখন Null Value Fill করব?**
*   **শর্ত:** যখন ডেটা Continuous Numerical (সংখ্যাতাত্ত্বিক) হয় এবং ডেটাতে **কোনো Outliers (অস্বাভাবিক মান) থাকে না**।
*   **লজিক:** ডেটা যদি নরমাল বা ব্যালেন্সড হয় (Normal Distribution), তখন গড় মানটি ডেটার সঠিক কেন্দ্র বা Central Tendency-কে প্রতিনিধিত্ব করে।
*   **উদাহরণ:** ধরি একটি ক্লাসের ছাত্রদের বয়সের কলামে কিছু ডেটা মিসিং। বাকি ছাত্রদের বয়স: `২১, ২২, ২৩, ২৪, ২২, ২৩`।
    এখানে কোনো আউটলায়ার নেই (যেমন কারো বয়স ৫ বা ৮০ নেই)। তাই এখানে Mean বা গড় (প্রায় ২২.৫) দিয়ে Null value ফিল করাটা সবচেয়ে লজিক্যাল এবং সেইফ।

#### **2. Median (মধ্যমা) দিয়ে কখন Null Value Fill করব?**
*   **শর্ত:** যখন ডেটা Numerical হয়, কিন্তু ডেটাতে **Outliers থাকে** বা ডেটা Skewed (একদিকে হেলানো) হয়।
*   **লজিক:** আমরা জানি, Mean আউটলায়ার দ্বারা মারাত্মকভাবে প্রভাবিত হয়। ডেটাতে খুব বড় বা খুব ছোট মান থাকলে Mean তার দিকে হেলে পড়ে এবং ভুল ইনসাইট দেয়। কিন্তু Median শুধু পজিশন বা মাঝখানের ভ্যালু নিয়ে কাজ করে। তাই আউটলায়ার থাকলেও এটি ডেটার সঠিক কেন্দ্র ধরে রাখতে পারে।
*   **উদাহরণ:** Salary বা ইনকামের কলাম। ধরি কয়েকজনের স্যালারি: `২০, ২৫, ২২, ২৪, ১০০০০ (Outlier)`।
    এখানে যদি Mean দিয়ে Null value ফিল করি, তবে সেটি প্রায় ২ হাজার বা তার বেশি হয়ে যাবে, যা সাধারণ স্যালারির সাথে মোটেও মেলে না! কিন্তু এই ডেটার Median (মাঝখানের মান) হবে `২৪`, যা সাধারণ ডেটার খুব কাছাকাছি। তাই আউটলায়ার থাকলে সবসময় **Median** ব্যবহার করতে হয়।

#### **3. Mode (প্রচুরক) দিয়ে কখন Null Value Fill করব?**
*   **শর্ত:** যখন ডেটা **Categorical (টেক্সট বা ক্যাটাগরি-ভিত্তিক)** হয়, অথবা Discrete (পূর্ণসংখ্যা) ডেটায় কোনো নির্দিষ্ট ভ্যালু সবচেয়ে বেশিবার রিপিট হয়।
*   **লজিক:** ক্যাটাগরিক্যাল ডেটার কোনো গড় (Mean) বা মধ্যমা (Median) বের করা গাণিতিকভাবে সম্ভব নয় (যেমন: 'Red' এবং 'Blue'-এর গড় হয় না)। তাই যে ক্যাটাগরিটি ডেটাসেটে সবচেয়ে বেশিবার উপস্থিত আছে (Most Frequent), সেটি দিয়ে আমরা খালি জায়গা পূরণ করে দিই।
*   **উদাহরণ ১ (Categorical):** T-shirt Color-এর কলামে কিছু ডেটা মিসিং। যদি দেখা যায় ডেটাসেটে 'Black' কালার সবচেয়ে বেশি আছে, তবে Null value গুলো 'Black' দিয়ে ফিল করা হয়।
*   **উদাহরণ ২ (Discrete Numerical):** পরিবারের সদস্য সংখ্যা। এটি ফ্র্যাকশন (যেমন ৩.৫ জন) হতে পারে না। যদি দেখা যায় বেশিরভাগ পরিবারে সদস্য সংখ্যা ৪, তাহলে মিসিং ভ্যালুগুলোতে Mode হিসেবে '৪' বসানো হয়।

---
**এক নজরে সামারি (Quick Summary):**
*   **Numerical Data + No Outliers** $\rightarrow$ Use **Mean**
*   **Numerical Data + Outliers Present** $\rightarrow$ Use **Median**
*   **Categorical Data / Text** $\rightarrow$ Use **Mode**

# **Understanding Variance in Data Preprocessing**

মেশিন লার্নিংয়ে ডেটার গুণগত মান (Quality) বোঝার জন্য Variance অত্যন্ত গুরুত্বপূর্ণ একটি পরিমাপ।

#### **1. The Problem with Mean (গড়ের সীমাবদ্ধতা)**
আমরা জানি, Mean বা গড় ডেটার সেন্ট্রাল পয়েন্ট বা কেন্দ্রীয় অবস্থা বোঝায়।
*   **Normal Case:** ধরি, স্টুডেন্টদের বয়সের ডেটা হলো: `১০, ১১, ১০, ১২, ১৩, ১১`।
    এখানে গড় বয়স হবে $\approx ১১.১$ বছর। এই গড়টি ডেটার সঠিক প্রতিনিধিত্ব করছে, কারণ সবাই ১১ বছরের আশেপাশেই আছে।
*   **Outlier Case:** কিন্তু ডেটাসেট যদি হয়: `১০, ১১, ১০, ১২, ১৩, ১১, ১০০০`।
    এখানে গড় বয়স হবে প্রায় **১৫১ বছর**! কিন্তু ক্লাসের কোনো স্টুডেন্টের বয়সই ১৫১ বছর নয়। এখানে `১০০০` হলো একটি চরম মান বা **Outlier**, যার কারণে পুরো গড়টি নষ্ট হয়ে গেছে।

#### **2. What is Variance? (Variance কী?)**
Variance আমাদের বলে দেয় ডেটাগুলো তাদের গড়ের চেয়ে ঠিক কতটা দূরে বা কতটা ছড়িয়ে-ছিটিয়ে (Spread) আছে। এটি বের করার জন্য আমরা প্রতিটি ডেটা পয়েন্ট থেকে গড়ের পার্থক্য বের করি এবং সেটিকে স্কয়ার (বর্গ) করে গড় করি।

**সূত্র:**
$$\text{Variance } (\sigma^2) = \frac{\sum (x - \mu)^2}{N}$$
*(যেখানে $x$ হলো প্রতিটি ডেটা, $\mu$ হলো গড়, এবং $N$ হলো মোট ডেটার সংখ্যা)*

#### **3. Calculating Variance (আপনার দেওয়া ডেটাসেট থেকে)**
আপনার দেওয়া ডেটাসেট: `০, ১১, ১০, ১২, ১৩, ১১, ১০০০`
এখানে মোট ডেটা ($N$) $= ৭$ টি।

**Step 1: Mean (গড়) বের করা**
$$\text{Sum} = ০ + ১১ + ১০ + ১২ + ১৩ + ১১ + ১০০০ = ১০৫৭$$
$$\text{Mean } (\mu) = \frac{১০৫৭}{৭} = ১৫১$$

**Step 2: প্রতিটি মান থেকে গড়ের পার্থক্যের স্কয়ার বের করা**
*   $(০ - ১৫১)^2 = ২২,৮০১$
*   $(১১ - ১৫১)^2 = ১৯,৬০০$
*   $(১০ - ১৫১)^2 = ১৯,৮৮১$
*   $(১২ - ১৫১)^2 = ১৯,৩২১$
*   $(১৩ - ১৫১)^2 = ১৯,০৪৪$
*   $(১১ - ১৫১)^2 = ১৯,৬০০$
*   $(১০০০ - ১৫১)^2 = ৭,২০,৮০১$ *(দেখুন, শুধুমাত্র এই একটি Outlier-এর কারণে মান কত বিশাল হয়ে গেছে!)*

**Step 3: Variance বের করা (সবগুলোর যোগফলকে $N$ দিয়ে ভাগ)**
$$\text{Sum of Squares} = ২২৮০১ + ১৯৬০০ + ১৯৮৮১ + ১৯৩২১ + ১৯০৪৪ + ১৯৬০০ + ৭২০৮০১ = ৮,৪১,০৪৮$$
$$\text{Variance } (\sigma^2) = \frac{৮,৪১,০৪৮}{৭} \approx ১,২০,১৪৯.৭১$$

#### **4. Interpretation in Machine Learning (এর মানে কী?)**

*   **High Variance (যেমন এখানে $১,২০,১৪৯$):**
    Variance এত বেশি হওয়ার মানে হলো এই ডেটাসেটে এমন কোনো ভ্যালু (Noise/Outlier) আছে, যা সাধারণ ডেটার চেয়ে অনেক দূরে। ডেটা এখানে Uniformly Distributed নয়। এই ডেটা সরাসরি মডেলে দিলে মডেল ঠিকমতো কাজ করবে না এবং ভুল প্যাটার্ন শিখবে।
*   **Low Variance:**
    যদি Variance খুব কম হতো (ধরা যাক ২ বা ৩), তাহলে আমরা নিশ্চিত হতাম যে ডেটাগুলো গড়ের খুব কাছাকাছি এবং সুন্দরভাবে সাজানো (Uniformly Distributed)। এখানে কোনো Noise নেই এবং মডেলটি এই ডেটা থেকে খুব সহজেই শিখতে পারবে।

**সিদ্ধান্ত:** Variance আমাদের ডেটার ভেতরের নয়েজ এবং আউটলায়ার শনাক্ত করতে সাহায্য করে, যাতে মডেল ট্রেইন করার আগে আমরা সেগুলো পরিষ্কার (Preprocess) করে নিতে পারি।



# **Understanding Standard Deviation (পরিমিত ব্যবধান)**

Standard Deviation (সংক্ষেপে SD বা $\sigma$) হলো ডেটাগুলো তাদের গড় (Mean) থেকে গড়ে কতটুকু দূরে ছড়ানো আছে, তার আসল এবং বাস্তবসম্মত পরিমাপ।

#### **1. The Fix: Variance থেকে Standard Deviation**
যেহেতু Variance বের করার সময় আমরা ডেটার পার্থক্যগুলোকে স্কয়ার (বর্গ) করে ফেলেছিলাম, তাই মানগুলো অস্বাভাবিক বড় হয়ে গিয়েছিল এবং তাদের একক (Unit) পরিবর্তন হয়ে গিয়েছিল।
এই মানটিকে আগের অবস্থায় (Original Unit) ফিরিয়ে আনার জন্য আমরা Variance-কে কেবল **Square Root (রুট)** করে দিই।

**সূত্র:**
$$\text{Standard Deviation } (\sigma) = \sqrt{\text{Variance}} = \sqrt{\frac{\sum (x - \mu)^2}{N}}$$

#### **2. Calculation (আগের ডেটাসেট থেকে)**
আমাদের আগের স্টুডেন্টদের বয়সের ডেটাসেটে:
*   গড় (Mean) ছিল $= ১৫১$ বছর
*   ভেদাঙ্ক (Variance) ছিল $= ১,২০,১৪৯.৭১$

এবার যদি আমরা এর Standard Deviation বের করি:
$$\text{SD } (\sigma) = \sqrt{১,২০,১৪৯.৭১} \approx ৩৪৬.৬২ \text{ বছর}$$
*(এর মানে হলো: এই ডেটাসেটে আউটলায়ার '১০০০' থাকার কারণে, ডেটাগুলো তাদের গড় ১৫১ বছর থেকে গড়ে প্রায় ৩৪৬ বছর এদিক-সেদিক ছড়ানো আছে!)*

#### **3. Variance এবং Standard Deviation-এর মধ্যে পার্থক্য**

| বিষয় | Variance (ভেদাঙ্ক) | Standard Deviation (পরিমিত ব্যবধান) |
| :--- | :--- | :--- |
| **গাণিতিক সম্পর্ক** | এটি Standard Deviation-এর বর্গ ($\sigma^2$) | এটি Variance-এর বর্গমূল বা রুট ($\sigma$) |
| **একক (Unit)** | এর একক ডেটার এককের বর্গ হয় (যেমন: $Age^2$, $TK^2$)। | এর একক মূল ডেটার এককের সমান থাকে (যেমন: $Age$, $TK$)। |
| **মান বা আকার** | মান অস্বাভাবিক বড় হয়ে যায় (স্কয়ার করার কারণে)। | মান বাস্তবসম্মত এবং ছোট থাকে। |
| **মেশিন লার্নিংয়ে ব্যবহার** | অ্যালগরিদমের ভেতরের গাণিতিক হিসাবে বা Cost Function-এ ব্যবহৃত হয়। | ডেটা স্কেলিং (Standardization / Z-score) এবং ডেটার বিস্তার বুঝতে সরাসরি ব্যবহৃত হয়। |
| **ব্যাখ্যা করা (Interpretability)** | মানুষের জন্য সরাসরি দেখে বোঝা কঠিন। | সরাসরি বোঝা যায় (গড় থেকে ডেটা কত দূরে আছে)। |

#### **4. Machine Learning-এ Standard Deviation-এর ম্যাজিক**
মেশিন লার্নিংয়ে আমরা যখন **Feature Scaling (Standardization)** করি, তখন আমরা মূলত Z-score ফর্মুলা ব্যবহার করি:
$$Z = \frac{x - \mu}{\sigma}$$
এখানে আমরা প্রতিটি ডেটা থেকে গড় বিয়োগ করে তাকে **Standard Deviation ($\sigma$)** দিয়ে ভাগ করি। এর ফলে বিশাল বড় বড় সংখ্যাগুলো (যেমন: স্যালারির ৫০,০০০ টাকা) ০ থেকে ১ এর মতো একটি ছোট স্কেলে চলে আসে, যা মেশিন লার্নিং মডেল খুব সহজেই প্রসেস করতে পারে।
