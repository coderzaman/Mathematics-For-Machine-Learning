## **Vectors in Machine Learning: Representing Data**

In machine learning, a **Vector** is simply a list of numbers. When we have a dataset, every single row (data point) is represented as a vector.

#### **1. Turning Data into Vectors**
Let's look at your dataset:
* **Study Hours (x)**
* **Play Hours (y)**
* **Result (Target)**

If a student studies for 7 hours and plays for 1 hour and passes, we represent their feature data as a 2-dimensional input vector:
**Input Vector:** $[7, 1]$
* The first dimension (x) is $7$ (Study).
* The second dimension (y) is $1$ (Play).

#### **2. Plotting the Vectors on a Graph**
Because our vectors have two dimensions ($x$ and $y$), we can visualize them as coordinates on a flat 2D plane (just like the image you provided).
* **Pass Class (Green):** $(7, 1)$ and $(6, 2)$
* **Fail Class (Red/Purple):** $(2, 5)$ and $(1, 6)$

#### **3. Classification using Vector Proximity (Distance)**
Now, imagine we get a completely new, unseen student:
* **New Data Point:** Study $= 4$, Play $= 1$.
* **New Vector:** $[4, 1]$
Screenshot from 2026-04-22 18-05-47.png

**How does the machine learning model classify this?**

If we plot $[4, 1]$ on the graph (the black dot in your image), the model calculates the physical distance between this new vector and all the other known vectors.
* Looking at the graph, the point $(4, 1)$ is physically much closer to the "Pass" vectors $(6, 2)$ and $(7, 1)$ than it is to the "Fail" vectors.
* **The Result:** Because its nearest neighbors belong to the "Pass" category, the model predicts that this new student will **Pass**.
* *Note: This specific logic is the foundation of a famous ML algorithm called **K-Nearest Neighbors (KNN)**.*

#### **4. Core Vector Operations in ML**
To make these predictions mathematically, models perform operations on these vectors. The most important ones are:
1.  **Vector Distance (Euclidean Distance):** Measures exactly how far apart two data points are on the graph. This is what we used in the example above to classify the new point!
2.  **Cosine Similarity:** Instead of measuring the distance between points, this measures the *angle* between two vectors. If they point in the exact same direction, they are highly similar. (This is heavily used in text analysis and recommendation systems).
3.  **Dot Product:** Multiplies vectors together to combine them. This is the fundamental calculation inside Neural Networks when multiplying inputs by weights ($w \cdot x$).

# **Scalars, Vectors, and Support Vector Machines (SVM)**



#### **1. Mathematical Definitions**
Before understanding advanced machine learning algorithms, it is important to distinguish between scalars and vectors:
* **Scalar:** A quantity represented *only* by its magnitude (size or amount). It has no directional component.
  * *Example:* Temperature ($30^\circ\text{C}$), or the length of a hill (30 meters).
* **Vector:** A quantity that requires both **magnitude** and **direction** to be represented correctly.
  * *Example:* Saying "the distance from A to B is 10 meters" is incomplete. Saying "A to B is 10 meters *towards the right*" makes it a vector.

#### **2. Vectors in Machine Learning Space**
image.png

In Machine Learning, we don't usually deal with physical directions like "North" or "Right." Instead, a vector represents a **data point in a multidimensional space**.
* Every feature in your dataset represents a new axis (dimension).
* By plotting these vectors in space, algorithms can examine their geometric relationships (like calculating the distance or angle between them) to make decisions and predictions.

#### **3. How Support Vector Machines (SVM) Work**
Support Vector Machines (SVM) is a powerful classification algorithm built entirely on this concept of vector geometry.

* **The Hyperplane:** SVM attempts to draw a definitive boundary to separate different classes of data (for example, drawing a line between the "Cats" and "Dogs" in our previous graph). In 2D space, this boundary is a flat line; in higher dimensions, it is called a **Hyperplane**.
* **The Support Vectors:** SVM is highly efficient because it does not care about every single data point. It only looks at the vectors (data points) that are *closest* to the dividing line. These critical points are called the **Support Vectors** because they literally "support" and define where the boundary is drawn.
* **Maximizing the Margin:** You could draw infinite lines to separate two groups. SVM's specific goal is to find the line that creates the **Maximum Margin**. The margin is the empty gap between the decision boundary and the nearest Support Vectors of each class. By making this "street" as wide as mathematically possible, the model becomes highly confident and robust when classifying new, unseen data.

# **Vectors and Matrices: Core Shapes and Operations**

#### **1. Vectors and Dimensions**
A vector can have any number of dimensions, which correspond to the number of features. The mathematical logic (like calculating the distance from the origin) works exactly the same across all dimensions.
* **2D Vector:** `[1, 2]`
* **3D Vector:** `[3, 2, 1]`
* **4D Vector:** `[1, 2, 3, 4]`
* **N-Dimensional Vector:** `[v1, v2, v3, ... vn]`

#### **2. Row vs. Column Vectors**
Vectors and matrices are defined by their shape, written as **(Rows $\times$ Columns)**.

* **Row Vector:** A vector where the number of rows is exactly **1**.
  * *Example:* $[v_1, v_2]$
  * *Shape:* $1 \times 2$ (1 row, 2 columns)

* **Column Vector:** A vector where the number of columns is exactly **1**.
  * *Example:*
    \begin{bmatrix}
    v_1 \\
    v_2 \\
    v_3
    \end{bmatrix}
  * *Shape:* $3 \times 1$ (3 rows, 1 column)

#### **3. Transpose Matrix ($T$)**
Transposing is the operation of flipping a matrix over its diagonal. Simply put: **Rows become columns, and columns become rows.**
* *Example:* Taking a $1 \times 3$ Row Vector and transposing it transforms it into a $3 \times 1$ Column Vector.
* $[v_1, v_2, v_3]^T \rightarrow$
$$
\begin{bmatrix}
v_1 \\
v_2 \\
v_3
\end{bmatrix}
$$

#### **4. Matrix Multiplication Rules**
When multiplying two matrices together (which happens constantly in neural networks), their dimensions must perfectly align.

* **The Golden Rule:** The number of **columns in the 1st matrix** MUST equal the number of **rows in the 2nd matrix**.
* **Valid Example:**
  * Matrix 1: $(2 \times 3)$
  * Matrix 2: $(3 \times 1)$
  * *Status:* **Valid!** The inner numbers match ($3 = 3$). The resulting new matrix will take the shape of the outer numbers: $(2 \times 1)$.
* **Invalid Example & The Fix:**
  * Matrix 1: $(2 \times 3)$
  * Matrix 2: $(1 \times 3)$
  * *Status:* **Invalid!** The inner numbers do not match ($3 \neq 1$).
  * *The Fix:* We can make this operation valid by applying a **Transpose** to Matrix 2. If we transpose $(1 \times 3)$, it flips into $(3 \times 1)$. Now, multiplying $(2 \times 3)$ by $(3 \times 1)$ works perfectly!

# **Vector Distance from the Origin (Magnitude)**


Screenshot from 2026-04-23 08-48-15.png
#### **1. What is it?**
The distance from the origin $(0,0)$ to a specific point in space represents the **Magnitude** or the **Length** of that vector. It tells us how far the data point is from the "zero state."

#### **2. Mathematical Calculation (Pythagoras Theorem)**
To find the length of a vector, we use the Pythagorean formula. Even if the direction changes (positive or negative), the distance remains the same because squaring a negative number results in a positive value.
Screenshot from 2026-04-23 09-04-42.png

* **2D Space:** For a point $(3, 2)$:
 $$\text{Distance} = \sqrt{3^2 + 2^2} = \sqrt{9 + 4} = \sqrt{13} \approx 3.61$$
* **Changing Direction:** For a point $(3, -2)$:
   $$\text{Distance} = \sqrt{3^2 + (-2)^2} = \sqrt{9 + 4} = 3.61$$

#### **3. Scaling to Higher Dimensions**
The beauty of this formula is that it stays the same regardless of how many features (dimensions) we have. In Machine Learning, this is called the **Euclidean Distance**.

* **3D Vector:** $\sqrt{x^2 + y^2 + z^2}$
* **N-Dimensional Vector:** $\sqrt{x_1^2 + x_2^2 + x_3^2 + \dots + x_n^2}$

---

### **Why do we measure this in Machine Learning?**

You mentioned being confused about how points "effect space." Here is the breakdown:

1.  **Effect on Space (Feature Importance):** In ML, if a vector has a very large magnitude, it means that specific data point has a "stronger" value. If we are calculating weights, a large distance from the origin suggests that the feature might have a significant impact on the model's decision.
2.  **Regularization (L2 Regularization):** When a model "overfits," it usually means the weights ($w$) have become too large (too far from the origin). We use the distance formula to calculate the size of these weights and then "penalize" the model if the distance is too high. This forces the model to keep weights small and simple.
3.  **Error Measurement:** When calculating the difference between a predicted value and an actual value, we are essentially measuring the "distance" between those two points in space.
4.  **Similarity (Cosine Similarity):** While distance tells us how **far** a point is, Cosine Similarity tells us the **angle**. By knowing the distance (magnitude), we can normalize vectors to see if they are pointing in the same direction, even if one is much longer than the other.


# **Distance Between Two Vectors (Euclidean Distance)**

In machine learning, calculating the distance between two vectors helps us understand how similar or different two data points are. This is universally calculated using the **Euclidean Distance** formula, which is derived directly from the Pythagorean theorem.

#### **1. Distance in 2D Space**

Screenshot from 2026-04-24 09-02-57.png

Let's take two vectors (data points):
* Vector $A = (2, 5)$
* Vector $B = (6, 3)$

If we plot these on a graph and draw a line between them, we can form a right-angle triangle.
* The horizontal distance (base) is the absolute difference of the x-coordinates: $|2 - 6| = 4$
* The vertical distance (height) is the absolute difference of the y-coordinates: $|5 - 3| = 2$

Because distance cannot be negative, we square these differences. The formula is:
$$\text{Distance} = \sqrt{(x_2 - x_1)^2 + (y_2 - y_1)^2}$$

Plugging in our numbers:
$$\text{Distance} = \sqrt{(6 - 2)^2 + (3 - 5)^2} = \sqrt{4^2 + (-2)^2} = \sqrt{16 + 4} = \sqrt{20} \approx 4.47$$

#### **2. Distance in 3D Space**
If we add a third feature to our dataset, our vectors become 3D: $A(x_1, y_1, z_1)$ and $B(x_2, y_2, z_2)$.
The geometry scales perfectly. We just add the third dimension to the formula:
$$\text{Distance} = \sqrt{(x_2 - x_1)^2 + (y_2 - y_1)^2 + (z_2 - z_1)^2}$$

#### **3. Distance in N-Dimensional Space**
In machine learning, datasets often have hundreds or thousands of features (N-dimensions).
Let $v_1 = (x_1, x_2, x_3, \dots, x_n)$ and $v_2 = (y_1, y_2, y_3, \dots, y_n)$.

Even though we cannot visualize a 100-dimensional triangle, the math remains exactly the same:
$$\text{Distance}_{ND} = \sqrt{(x_1 - y_1)^2 + (x_2 - y_2)^2 + (x_3 - y_3)^2 + \dots + (x_n - y_n)^2}$$



# **The Dot Product of Two Vectors**

The dot product is one of the most important vector operations in Machine Learning. Instead of returning a new vector, the dot product multiplies two vectors together to return a **single scalar number**.

#### **1. Applications in Machine Learning**
* **Vector Projection:** It helps us find the "shadow" or projection of one vector directly onto another.
* **Dimensionality Reduction (PCA):** Principal Component Analysis (PCA) uses projections to combine and reduce the number of features in a massive dataset while keeping the most important information.
* **Cosine Similarity:** By combining the dot product with the vector distance (magnitude), we can calculate the exact **Angular Distance** between two vectors to see how similar they are.

#### **2. The Mathematical Calculation**
To calculate the dot product, we multiply the matching dimensions of each vector and sum the results.
* Let Vector $A = [1, 2, 4]$
* Let Vector $B = [2, 3, 5]$

$$A \cdot B = (x_1 \times y_1) + (x_2 \times y_2) + (x_3 \times y_3)$$
$$A \cdot B = (1 \times 2) + (2 \times 3) + (4 \times 5)$$
$$A \cdot B = 2 + 6 + 20 = 28$$

#### **3. Matrix Multiplication Rules & Transposition**
For the computer to calculate this, it must follow strict matrix multiplication rules.
If $A$ and $B$ are both stored as **Column Vectors** (shape $3 \times 1$), we cannot multiply them directly because the inner dimensions do not match: $(3 \times 1) \times (3 \times 1)$ is invalid.

To fix this, we apply a **Transpose** to the first vector:
* $A^T$ becomes a **Row Vector** with shape $(1 \times 3)$.
* Now, $A^T B$ becomes $(1 \times 3) \times (3 \times 1)$.
* This is valid! The result is a $(1 \times 1)$ matrix, which is our scalar answer ($28$).

#### **4. The Machine Learning Equation**
This exact matrix multiplication rule is why the fundamental equation for linear layers and neural networks is written the way it is.

Let's say we have 3 features ($x_1, x_2, x_3$) and 3 weights ($w_1, w_2, w_3$).
* Data Vector $X = [x_1, x_2, x_3]$
* Weight Vector $W = [w_1, w_2, w_3]$

Instead of writing out $y = w_1x_1 + w_2x_2 + w_3x_3 + b$, we represent it cleanly using the dot product:
**$$y = W^T X + b$$**

### **Vector Operations: Scalars and Vectors**

#### **1. Scalar Operations with Vectors (Shifting)**
When we perform operations (like addition or subtraction) between a vector and a single scalar value, that scalar is applied to *every single component* of the vector independently. Geometrically, this shifts the vector's position in space.

* **Addition:** $$A_{\text{vector}} = [2, 1] + 3 = [2+3, 1+3] = [5, 4]$$
Screenshot from 2026-04-24 11-11-30.png

* **Subtraction:** $$A_{\text{vector}} = [5, 4] - 3 = [5-3, 4-3] = [2, 1]$$

Screenshot from 2026-04-24 11-15-29.png

* **Machine Learning Context (Mean Centering):** This scalar shifting is heavily used in preprocessing data. If we calculate the mean (average) of a feature and subtract that scalar value from every data point in the vector, we "shift" the entire dataset so it is centered perfectly around zero.

#### **2. Vector Operations with Vectors**
To add or subtract two vectors together, they **must have the exact same shape** (the same number of dimensions).

##### **A. Vector Addition**

Screenshot from 2026-04-24 11-58-45.png

Mathematically, we add the matching dimensions together ($x$ with $x$, $y$ with $y$).
* $\vec{V} = [2, 3]$
* $\vec{A} = [1, 2]$
* $$\vec{V} + \vec{A} = [2+1, 3+2] = [3, 5]$$

**What happens geometrically? (The Tail-to-Head Rule)**
Imagine walking! You walk along the path of Vector $V$. Once you reach the tip (head) of Vector $V$, you place the tail of Vector $A$ right there and walk along its path. The final result is a brand new vector drawn from your starting point (the origin) directly to your final destination.

##### **B. Vector Subtraction**
Screenshot from 2026-04-24 12-04-19.png
Technically, vectors do not have a pure "subtract" operation. Just like you noted, subtraction is actually just **adding the opposite** vector.
* $\vec{V} = [8, 3]$
* $\vec{A} = [3, 1]$

Instead of subtracting, we flip the direction of the second vector (making it $-\vec{A}$) and then use the exact same Tail-to-Head addition rule:
* $$\vec{V} - \vec{A} = \vec{V} + (-\vec{A})$$
* $$\vec{V} + (-\vec{A}) = [8, 3] + [-3, -1] = [5, 2]$$
