# **Matrices in Machine Learning and Deep Learning**

In Linear Algebra, a matrix is a rectangular array of numbers arranged in rows and columns. In Machine Learning (ML) and Deep Learning (DL), matrices are the absolute foundation of how we store data and perform calculations.

#### **1. Why Do We Need Matrices? (Data Representation)**
Most real-world datasets are tabular. A matrix perfectly mirrors this structure, allowing computers to read and process the entire dataset as a single mathematical object.

* **Columns = Features & Targets:** The inputs (what the model learns from) are the features, and the output (what we want to predict) is the target.
* **Rows = Data Points:** Each row represents one individual observation (e.g., a single student).

**Example 1: Single Feature Dataset**
| Study Hours (Input Feature) | Marks (Target Output) |
| :--- | :--- |
| 3 | 5 |
| 5 | 20 |
| 6 | 35 |

**Example 2: Multi-Feature Dataset**
| Study Hours (Feature 1) | Play Hours (Feature 2) | Marks (Target Output) |
| :--- | :--- | :--- |
| 4 | 2 | 25 |
| 2 | 5 | 10 |

#### **2. The Core ML Logic (Linear Regression in 2D)**
When we feed this data into a model, the model tries to build a mathematical logic. For 2D data (one feature and one target), the model attempts to draw a "Best Fit Line" through the data points.

Screenshot from 2026-05-02 16-38-19.png

The equation for this prediction line is:
$$Prediction = (\text{Weight} \times \text{Feature}) + \text{Bias}$$
$$y_{pre} = wx + b$$

By adjusting the weight (slope) and bias, the model tries to align the line as perfectly as possible with the actual data points. Once the line is set, any new, unseen data can be predicted using this line.

#### **3. The Calculation Problem: Scaling Up**
Imagine you have 100 data points. If you calculate them one by one using a standard loop:
* $y_1 = w_1 \times x_1$
* $y_2 = w_1 \times x_2$
* ...
* $y_{100} = w_1 \times x_{100}$

Now imagine you have multiple features: $y = w_1x_1 + w_2x_2 + w_3x_3 + w_4x_4 + b$.
Doing this one by one for millions of data points requires an immense amount of looping, which is incredibly slow.

#### **4. The Solution: Matrix Multiplication & Hardware Optimization**
Instead of calculating row by row, we group the weights into a **Weight Vector** and the features into a **Feature Matrix**.

$$
\begin{bmatrix}
w_1 \\
w_2 \\
w_3 \\
w_4
\end{bmatrix}
\times
\begin{bmatrix}
f_{1,1} & f_{1,2} & f_{1,3} & \dots \\
f_{2,1} & f_{2,2} & f_{2,3} & \dots \\
f_{3,1} & f_{3,2} & f_{3,3} & \dots \\
f_{4,1} & f_{4,2} & f_{4,3} & \dots
\end{bmatrix}
$$

**Why is this better?**
1. **SIMD (Single Instruction, Multiple Data):** Modern CPUs are highly optimized for matrix multiplication. Instead of executing 100 different multiply instructions, the CPU uses SIMD to load multiple data points into memory at once and applies a single multiplication instruction to all of them simultaneously.
2. **GPU Parallelism:** Graphics Processing Units (GPUs) are explicitly designed to perform thousands of matrix operations at the exact same time (parallelism). By formatting our ML equations as matrices, we unlock the massive, parallel computing power of the GPU, turning tasks that would take days into tasks that take seconds.

# **The Matrix Skeleton: Scalars, Vectors, and Matrices**

To understand how data is structured in Machine Learning, we need to understand the hierarchy of mathematical shapes.

#### **1. The Core Components**
* **Scalar:** A single value with no direction (0 dimensions).
  * *Example:* `15`
* **Vector:** A 1-dimensional array of numbers. It can be horizontal or vertical.
  * **Row Vector:** $[10, 20, 30]$
  * **Column Vector:**
    
    \begin{bmatrix}
    10 \\
    20 \\
    30
    \end{bmatrix}

* **Matrix:** A 2-dimensional grid of numbers, built by stacking multiple vectors together.
  * *Example:*

    \begin{bmatrix}
    10 & 20 & 30 \\
    20 & 30 & 40 \\
    50 & 60 & 70
    \end{bmatrix}


#### **2. Matrix Shape and Dimensions**
A matrix is defined by its "shape," which is always written as **Rows $\times$ Columns**.
* Let's look at this matrix:
  $$
  \begin{bmatrix}
    10 & 20 & 30 \\
    20 & 30 & 40
  \end{bmatrix}
  $$
* It has **2 rows** and **3 columns**.
* **Shape:** $2 \times 3$ matrix.
* **Total Elements:** Simply multiply the rows by columns ($2 \times 3 = 6$ elements).

#### **3. The Machine Learning Perspective**
In ML, matrices are not just abstract numbers; they are the exact blueprint of your dataset.

* **Rows = Data Points:** Every individual row represents a unique sample (e.g., a specific student, a single image, one house).
* **Columns = Features:** Every column represents an independent variable or attribute we are using to make predictions (e.g., study hours, play hours).

**Example:**
We have 3 students (3 Data Points) and 2 variables (2 Features: Study Hours, Play Hours).

| Study Hour | Play Hour |
| :--- | :--- |
| 8 | 2 |
| 7 | 4 |
| 7.5 | 3 |

**As a Matrix:**
$$
\begin{bmatrix}
8 & 2 \\
7 & 4 \\
7.5 & 3
\end{bmatrix}
$$

* **Dataset Shape:** This dataset has a shape of **$3 \times 2$**.
* If we download a massive dataset with a shape of **$200 \times 3$**, we immediately know we are looking at exactly $200$ data points (rows), and each point is described by $3$ different features (columns).


# **Classification of Matrices in Machine Learning**

Different shapes of matrices serve distinct roles in machine learning algorithms. Here are the core classifications:

#### **1. Row Matrix (Row Vector)**
A matrix with only **1 row** and multiple columns.
* **Shape:** $(1 \times n)$ dimension.
* **Example:** $[10, 20, 30 \dots n]$
* **Machine Learning Context:**
  * A single data point in a dataset (e.g., one specific student's Study, Play, and IQ scores).
  * When you feed **new data** into a model to get a prediction, you input it as a row vector.

#### **2. Column Matrix (Column Vector)**
A matrix with only **1 column** and multiple rows.
* **Shape:** $(n \times 1)$ dimension.
* **Example:**
  $$
  \begin{bmatrix}
  10 \\
  30 \\
  40
  \end{bmatrix}
  $$
* **Machine Learning Context:**
  * The **Target Column** (the output/labels you are trying to predict).
  * The **Weight Matrix** ($W$) holding the model's parameters.
  * The final **Prediction Matrix** generated by the model.

#### **3. Square Matrix**
A matrix where the **number of rows perfectly equals the number of columns**.
* **Shape:** $(n \times n)$ dimension.
* **Example:** Let Matrix $A$ be a $3 \times 3$ square matrix:
  $$
  A =
  \begin{bmatrix}
  10 & 20 & 30 \\
  30 & 40 & 50 \\
  50 & 60 & 70
  \end{bmatrix}
  $$
* **Machine Learning Context:** In Principal Component Analysis (PCA), the **Covariance Matrix** is always a square matrix used to find relationships between features.

**Diagonals & Notation in a Square Matrix:**
* **Primary Diagonal:** Runs from top-left to bottom-right (Values: `10, 40, 70`).
* **Secondary Diagonal:** Runs from top-right to bottom-left (Values: `30, 40, 50`).
* **Position Notation ($A_{ij}$):** $i$ represents the Row, $j$ represents the Column.
  * *Example:* $A_{31}$ means Row 3, Column 1. In our matrix, $A_{31} = 50$.

#### **4. Identity Matrix ($I$)**
A special type of Square Matrix where the **primary diagonal values are all $1$**, and every other value is $0$.
* **Example ($3 \times 3$ Identity):**
  $$
  I =
  \begin{bmatrix}
  1 & 0 & 0 \\
  0 & 1 & 0 \\
  0 & 0 & 1
  \end{bmatrix}
  $$
* **The Golden Rule:** If you multiply ANY matrix by an Identity Matrix, you get the exact same matrix back (just like multiplying a number by $1$).
  * **Equation:** $A \times I = A$

#### **5. Transpose Matrix ($A^T$)**
The operation of flipping a matrix over its diagonal. **Rows become columns, and columns become rows.**
* **Example:** Transforming a $3 \times 2$ matrix into a $2 \times 3$ matrix.
  $$
  \begin{bmatrix}
  10 & 20 \\
  40 & 50 \\
  60 & 70
  \end{bmatrix}^T
  \rightarrow
  \begin{bmatrix}
  10 & 40 & 60 \\
  20 & 50 & 70
  \end{bmatrix}
  $$


# **Matrix Addition, Subtraction, and Broadcasting in ML**

#### **1. Standard Matrix Addition and Subtraction**
For standard operations, both matrices must have the exact same dimensions (shape). The operation happens **element-wise** (matching positions are added/subtracted). The resulting matrix will also have the exact same shape.

Let Matrix $A$ and Matrix $B$ be $2 \times 3$ matrices:
$$A = \begin{bmatrix} 10 & 20 & 30 \\ 40 & 50 & 60 \end{bmatrix}, \quad B = \begin{bmatrix} 1 & 2 & 3 \\ 4 & 5 & 6 \end{bmatrix}$$

*   **Addition ($A + B$):**
    A + B = \begin{bmatrix} 10+1 & 20+2 & 30+3 \\ 40+4 & 50+5 & 60+6 \end{bmatrix} = \begin{bmatrix} 11 & 22 & 33 \\ 44 & 55 & 66 \end{bmatrix}
*   **Subtraction ($A - B$):**
    A - B = \begin{bmatrix} 10-1 & 20-2 & 30-3 \\ 40-4 & 50-5 & 60-6 \end{bmatrix} = \begin{bmatrix} 9 & 18 & 27 \\ 36 & 45 & 54 \end{bmatrix}

---

#### **2. Broadcasting: Scalars (0D to 2D)**
While strict algebra forbids adding a scalar to a matrix, ML libraries (NumPy/PyTorch) use **Broadcasting**. They automatically "stretch" the scalar into a matrix of the same shape to make the operation possible.

*   **Addition ($A + 10$):** Stretches $10$ to a $2 \times 3$ matrix of tens.
    \begin{bmatrix} 10 & 20 \\ 30 & 40 \end{bmatrix} + \begin{bmatrix} 10 & 10 \\ 10 & 10 \end{bmatrix} = \begin{bmatrix} 20 & 30 \\ 40 & 50 \end{bmatrix}
*   **Subtraction ($A - 10$):**
    \begin{bmatrix} 10 & 20 \\ 30 & 40 \end{bmatrix} - \begin{bmatrix} 10 & 10 \\ 10 & 10 \end{bmatrix} = \begin{bmatrix} 0 & 10 \\ 20 & 30 \end{bmatrix}

---

#### **3. Broadcasting: Row Vectors (1D to 2D)**
To broadcast a Row Vector into a Matrix, the **number of columns must match**. The vector is "stretched" vertically by duplicating its rows.

Let $A = \begin{bmatrix} 10 & 20 & 30 \\ 40 & 50 & 60 \end{bmatrix}$ and $B_{\text{row}} = [1, 2, 3]$.

*   **Addition ($A + B_{\text{row}}$):** The vector stretches downwards.
    $$\begin{bmatrix} 10 & 20 & 30 \\ 40 & 50 & 60 \end{bmatrix} + \begin{bmatrix} 1 & 2 & 3 \\ 1 & 2 & 3 \end{bmatrix} = \begin{bmatrix} 11 & 22 & 33 \\ 41 & 52 & 63 \end{bmatrix}$$
*   **Subtraction ($A - B_{\text{row}}$):**
    $$\begin{bmatrix} 10 & 20 & 30 \\ 40 & 50 & 60 \end{bmatrix} - \begin{bmatrix} 1 & 2 & 3 \\ 1 & 2 & 3 \end{bmatrix} = \begin{bmatrix} 9 & 18 & 27 \\ 39 & 48 & 57 \end{bmatrix}$$
*   *Note on failure:* If $B = [1, 2, 3, 4]$, it cannot be broadcast because the columns ($4$) do not match Matrix A's columns ($3$).

---

#### **4. Broadcasting: Column Vectors (1D to 2D)**
To broadcast a Column Vector into a Matrix, the **number of rows must match**. The vector is "stretched" horizontally by duplicating its columns.

Let $A = \begin{bmatrix} 10 & 20 & 30 \\ 40 & 50 & 60 \end{bmatrix}$ and $B_{\text{col}} = \begin{bmatrix} 1 \\ 2 \end{bmatrix}$.

*   **Addition ($A + B_{\text{col}}$):** The vector stretches to the right.
    $$\begin{bmatrix} 10 & 20 & 30 \\ 40 & 50 & 60 \end{bmatrix} + \begin{bmatrix} 1 & 1 & 1 \\ 2 & 2 & 2 \end{bmatrix} = \begin{bmatrix} 11 & 21 & 31 \\ 42 & 52 & 62 \end{bmatrix}$$
*   **Subtraction ($A - B_{\text{col}}$):**
    $$\begin{bmatrix} 10 & 20 & 30 \\ 40 & 50 & 60 \end{bmatrix} - \begin{bmatrix} 1 & 1 & 1 \\ 2 & 2 & 2 \end{bmatrix} = \begin{bmatrix} 9 & 19 & 29 \\ 38 & 48 & 58 \end{bmatrix}$$
*   *Note on failure:* If $B = \begin{bmatrix} 1 \\ 2 \\ 3 \end{bmatrix}$, it cannot be broadcast because the rows ($3$) do not match Matrix A's rows ($2$).

---

### **Why Do We Need This in Machine Learning?**
1.  **Adding Bias ($Y = WX + b$):** When processing a massive batch of 1,000 data points through a neural network layer, the output is a matrix. However, the bias ($b$) is usually just a single row vector. Broadcasting automatically adds that bias vector to every single row in the batch simultaneously!
2.  **Gradient Descent Updates:** When we update weights using $W_{\text{new}} = W_{\text{old}} - (\text{learning_rate} \times \text{gradient})$, we are using scalar multiplication and matrix subtraction to update millions of parameters at once.

# **Matrix Multiplication in Machine Learning**



Matrix multiplication is the mathematical engine that powers almost all modern Deep Learning and AI. Unlike addition, it does not happen element-by-element. Instead, it uses the **Dot Product** of rows and columns.

#### **1. The Golden Rule of Dimensions**
To multiply Matrix $A$ by Matrix $B$, the **number of columns in $A$ MUST equal the number of rows in $B$**.
*   Let $A$ have shape $(n, m)$
*   Let $B$ have shape $(x, y)$
*   **Condition:** $m$ must equal $x$.
*   **Result Shape:** The resulting matrix will take the outer dimensions: $(n, y)$.

**Examples:**
*   $A(2, 3) \times B(3, 5) \rightarrow$ **Valid!** Result is $(2, 5)$.
*   $A(2, 3) \times B(5, 7) \rightarrow$ **Invalid!** ($3 \neq 5$).

#### **2. How the Math Works (Row $\times$ Column)**
When multiplying $A \times B$, we take a **Row** from $A$ and calculate the dot product with a **Column** from $B$.

Let $A = \begin{bmatrix} 1 & 2 & 3 \\ 4 & 5 & 6 \end{bmatrix}$ (Shape: $2 \times 3$)
Let $B = \begin{bmatrix} 10 & 7 \\ 20 & 8 \\ 30 & 9 \end{bmatrix}$ (Shape: $3 \times 2$)

Because inner dimensions match ($3 = 3$), our result will be a $2 \times 2$ matrix.

**Step-by-Step Calculation:**
*   **Top-Left Cell (Row 1 $\times$ Col 1):**
    $(1 \times 10) + (2 \times 20) + (3 \times 30) = 10 + 40 + 90 = \mathbf{140}$
*   **Top-Right Cell (Row 1 $\times$ Col 2):**
    $(1 \times 7) + (2 \times 8) + (3 \times 9) = 7 + 16 + 27 = \mathbf{50}$
*   **Bottom-Left Cell (Row 2 $\times$ Col 1):**
    $(4 \times 10) + (5 \times 20) + (6 \times 30) = 40 + 100 + 180 = \mathbf{320}$
*   **Bottom-Right Cell (Row 2 $\times$ Col 2):**
    $(4 \times 7) + (5 \times 8) + (6 \times 9) = 28 + 40 + 54 = \mathbf{122}$

**Final Result Matrix:**
$$A \times B = \begin{bmatrix} 140 & 50 \\ 320 & 122 \end{bmatrix}$$

---

#### **3. The Machine Learning Perspective: Why do we need this?**

In Machine Learning, we don't multiply matrices just for fun. We do it to process massive amounts of data instantly.

**A. Neural Network Layers (The Forward Pass)**
Imagine you have a dataset of 1,000 students. Each student has 3 features (Study, Play, Sleep).
*   Your **Input Data Matrix ($X$)** has a shape of $(1000 \times 3)$.
*   Your model has a **Weight Matrix ($W$)** mapping those 3 features to 5 hidden "neurons". Its shape is $(3 \times 5)$.

Instead of writing a complex loop to calculate the math for every single student, you just write:
**$X \times W$**
Because $(1000 \times 3) \times (3 \times 5)$ is valid, the GPU calculates all millions of combinations simultaneously and outputs a $(1000 \times 5)$ matrix. Matrix multiplication allows ML to process entire datasets in milliseconds.

**B. Data Transformations**
Whenever you need to change the dimensionality of data—like reducing 100 features down to 2 features so you can draw it on a graph (PCA)—you multiply your dataset by a specially crafted Transformation Matrix.
