# **Intro to Probability (প্রোবাবিলিটি বা সম্ভাব্যতা)**

Probability বা সম্ভাব্যতা দিয়ে আমরা পরিমাপ করি কোনো একটি ঘটনা ঘটার সম্ভাবনা ঠিক কতটুকু। একটি ঘটনা ঘটার পেছনে অনেকগুলো প্রভাবক (Influence) কাজ করতে পারে।

সাধারণত Probability-এর মান গাণিতিকভাবে **০ থেকে ১** এর মধ্যে থাকে।

* **$0 \rightarrow$** ঘটনাটি ঘটার কোনো সম্ভাবনাই নেই (Impossible Event)।
* **$0.5 \rightarrow$** ঘটনাটি ঘটার এবং না ঘটার সম্ভাবনা একদম সমান (50/50 Chance)।
* **$1 \rightarrow$** ঘটনাটি ঘটার সম্ভাবনা ১০০% নিশ্চিত (Certain Event)।

image.png

#### **Examples (বাস্তব উদাহরণ):**

* **Coin Toss (কয়েন টস):** একটি কয়েন টস করলে আমাদের Sample Space বা সম্ভাব্য ফলাফল ২ টি: $S = \{H, T\}$। এখান থেকে Head পড়ার সম্ভাবনা $0.5$ (বা ৫০%)।
* **Rolling a Die (ছক্কা নিক্ষেপ):** একটি ছক্কায় Sample Space ৬ টি: $S = \{1, 2, 3, 4, 5, 6\}$। এখান থেকে যেকোনো একটি নির্দিষ্ট সংখ্যা (যেমন ৬) ওঠার সম্ভাবনা হলো $\frac{1}{6}$।

### **Machine Learning Aspect (মেশিন লার্নিংয়ে প্রোবাবিলিটি)**

মেশিন লার্নিংয়ের ক্লাসিফিকেশন (Classification) মডেলগুলো মূলত প্রোবাবিলিটি বা সম্ভাব্যতার ওপর ভিত্তি করেই কাজ করে।

**উদাহরণ (Image Classifier):**
আমরা যদি একটি বিড়াল বা কুকুরের ছবি ক্লাসিফাই করার মডেল কল্পনা করি, তখন মডেলটি সরাসরি "এটি বিড়াল" না বলে প্রোবাবিলিটি ভ্যালু আউটপুট দেয়:
`Image` $\rightarrow$ `Model` $\rightarrow$ `[0.8 Cat, 0.2 Dog]`
যেহেতু এখানে বিড়াল হওয়ার প্রোবাবিলিটি (০.৮) বেশি, তাই মডেল সিদ্ধান্ত নেয় যে ছবিটি একটি বিড়ালের।

### **Logistic Regression (লজিস্টিক রিগ্রেশন)**

লজিস্টিক রিগ্রেশন একটি বাইনারি ক্লাসিফিকেশন (Binary Classification) অ্যালগরিদম, যা Yes/No বা 0/1 আকারে রেজাল্ট দেয়। এটি ডেটাকে প্রেডিক্ট করার জন্য প্রোবাবিলিটি ব্যবহার করে।

**Dataset Example:**

| Study Hour ($x_1$) | Play Hour ($x_2$) | Pass ($y$) |
| --- | --- | --- |
| 8 | 2 | Yes (1) |
| 7 | 3 | Yes (1) |
| 5 | 4 | Yes (1) |
| 4 | 7 | No (0) |
| 3 | 6 | No (0) |

**How it works:**
আমরা যখন নতুন কোনো আনসিন (Unseen) ডেটা পয়েন্ট মডেলে ইনপুট দিই (ধরি, `Study = 2, Play = 7`), তখন মডেলটি লিনিয়ার রিগ্রেশনের মতো একটি সমীকরণ ($z = w_1x_1 + w_2x_2 + b$) তৈরি করে।

কিন্তু ক্লাসিফিকেশনের জন্য মডেল এই মানটিকে **Sigmoid Function**-এর ভেতর দিয়ে পাস করায়।
**Sigmoid Function:** $\sigma(z) = \frac{1}{1 + e^{-z}}$
image.png
সিগময়েড ফাংশনের কাজ হলো যেকোনো বড় বা ছোট সংখ্যাকে ০ থেকে ১ এর মাঝখানে (Probability value) নিয়ে আসা।

**Decision Boundary (থ্রেশহোল্ড):**
লজিস্টিক রিগ্রেশন সাধারণত $0.5$-কে থ্রেশহোল্ড হিসেবে ধরে।

* সিগময়েড থেকে পাওয়া মান যদি $\ge 0.5$ হয়, মডেল প্রেডিক্ট করে **Yes (1)**।
* মানটি যদি $< 0.5$ হয়, মডেল প্রেডিক্ট করে **No (0)**।

আপনার উদাহরণ অনুযায়ী, যদি সিগময়েড ফাংশন থেকে প্রোবাবিলিটি আসে **০.২**, যেহেতু এটি ০.৫ এর চেয়ে কম, তাই মডেল এই স্টুডেন্টের জন্য প্রেডিক্ট করবে **No বা Fail**।

### **Naive Bayes (নেইভ বেইজ)**

এটি আরেকটি পপুলার ক্লাসিফিকেশন অ্যালগরিদম, যা সম্পূর্ণভাবে প্রোবাবিলিটির **Bayes' Theorem**-এর ওপর নির্ভর করে কাজ করে। এটি প্রতিটি ফিচারের আলাদা আলাদা প্রোবাবিলিটি হিসাব করে চূড়ান্ত সিদ্ধান্ত নেয় (যেমন: ইমেইল স্প্যাম নাকি স্প্যাম নয় তা বের করতে)।

# **General Formula of Probability (সম্ভাব্যতার সাধারণ সূত্র)**

প্রোবাবিলিটি বা সম্ভাব্যতা বের করার সবচেয়ে বেসিক এবং মূল সূত্রটি হলো:

$$P(\text{Event}) = \frac{\text{Favorable Outcomes}}{\text{Total Possible Outcomes}}$$

**গুরুত্বপূর্ণ টার্মগুলো (Key Terms):**

* **Sample Space ($S$):** একটি ঘটনায় যত ধরনের ফলাফল আসা সম্ভব, তার সম্পূর্ণ সেট (Total Possible Outcomes)।
* **Event:** আমরা যে নির্দিষ্ট ফলাফলটি চাচ্ছি (Favorable Outcome)।

#### **Example 1: Coin Toss (কয়েন টস)**

একটি কয়েন ফ্লিপ করলে Head অথবা Tail পড়বে।

* **Total Possible Outcome ($S$):** $\{H, T\}$ (মোট ফলাফল ২ টি)
* আমরা বের করতে চাই Head পড়ার সম্ভাব্যতা।
* **Favorable Outcome:** $1$ (কারণ ১টি হেড আছে)

**Calculation:**


$$P(Head) = \frac{1}{2} = 0.5$$


অর্থাৎ, Head পড়ার সম্ভাবনা **50%** এবং বাকি **50%** সম্ভাবনা আছে Tail পড়ার।

#### **Example 2: Single Die Roll (একটি ছক্কা নিক্ষেপ)**

একটি ছক্কা নিক্ষেপ করলে ১ থেকে ৬ পর্যন্ত যেকোনো একটি সংখ্যা আসতে পারে।

* **Sample Space ($S$):** $\{1, 2, 3, 4, 5, 6\}$ (মোট ফলাফল ৬ টি)
* যদি বলা হয়, ছক্কায় ঠিক $4$ আসার সম্ভাব্যতা কত? এখানে $4$ আছে মাত্র ১ বার।

**Calculation:**


$$P(4) = \frac{1}{6} \approx 0.167$$


ছক্কার প্রতিটি পিঠ পড়ার সম্ভাবনাই সমান (Uniform Probability)। তাই:


$$P(1) = P(2) = P(3) = P(4) = P(5) = P(6) \approx 0.167$$

#### **Example 3: Conditional Die Roll (শর্তযুক্ত সম্ভাব্যতা)**

এবার যদি প্রশ্ন করা হয়, ছক্কায় ৩ বা তার সমান/বড় ( $\ge 3$ ) সংখ্যা আসার সম্ভাব্যতা কত?

* **Sample Space ($S$):** $\{1, 2, 3, 4, 5, 6\}$ (মোট ফলাফল ৬ টি)
* **Favorable Outcome:** $\{3, 4, 5, 6\}$ (শর্ত অনুযায়ী আমাদের অনুকূল ফলাফল ৪ টি)

**Calculation ( $\ge 3$ আসার সম্ভাবনা):**


$$P(\ge 3) = \frac{4}{6} \approx 0.67 \text{ (বা 67\%)}$$

**Calculation ( $3$ এর কম আসার সম্ভাবনা):**
বাকি যে সংখ্যাগুলো রইল, অর্থাৎ $\{1, 2\}$, সেগুলো আসার সম্ভাবনা:


$$P(< 3) = \frac{2}{6} \approx 0.33 \text{ (বা 33\%)}$$

# **Probability Tree (সম্ভাব্যতা বৃক্ষ)**

যখন একাধিক ঘটনা পরপর ঘটে (Sequential Events) এবং সম্ভাব্য ফলাফলগুলো (Sample Space) সাধারণ নিয়মে বের করা একটু জটিল বা কনফিউজিং হয়ে যায়, তখন আমরা সেগুলোকে সহজে ভিজ্যুয়ালাইজ করার জন্য **Probability Tree** ব্যবহার করি।

গাছের ডালপালার মতো ধাপে ধাপে প্রতিটি ইভেন্টকে ভেঙে লেখা হয় বলে একে ট্রি বলা হয়।

#### **Example 1: Tossing 2 Coins (২টি কয়েন পরপর ফ্লিপ করা)**

প্রথম কয়েন ফ্লিপ করলে আসবে Head ($H$) বা Tail ($T$)। এরপর দ্বিতীয় কয়েন ফ্লিপ করলে প্রথমটির প্রতিটি ফলাফলের শাখা থেকে আবার নতুন করে $H$ বা $T$ আসবে।

Screenshot from 2026-05-19 20-30-31.png

* **Total Outcomes / Sample Space ($S$):** $\{HH, HT, TH, TT\}$ (মোট ৪টি সম্ভাব্য ফলাফল)
* **প্রশ্ন:** পরপর ২টি Head আসার সম্ভাব্যতা কত?
* **Favorable Outcome:** $\{HH\}$ (মাত্র ১ বার আছে)

**Calculation:**


$$P(H, H) = \frac{1}{4} = 0.25$$


*(অর্থাৎ, ২৫% সম্ভাবনা আছে পরপর দুটো হেড পড়ার)*

#### **Example 2: Rolling a Die and Tossing a Coin (১টি ছক্কা এবং ১টি কয়েন)**

এটি আগেরটির চেয়ে আরেকটু কমপ্লেক্স উদাহরণ। প্রথমে একটি ছক্কা মারলে ফলাফল আসবে ১ থেকে ৬। এরপর কয়েন ফ্লিপ করলে প্রতিটি সংখ্যার শাখা থেকে $H$ বা $T$ যুক্ত হবে।

Screenshot from 2026-05-19 20-37-11.png

* **Total Outcomes / Sample Space ($S$):** $\{1H, 1T, 2H, 2T, 3H, 3T, 4H, 4T, 5H, 5T, 6H, 6T\}$ (মোট ১২টি সম্ভাব্য ফলাফল)
* **প্রশ্ন:** ছক্কায় জোড় সংখ্যা (Even) এবং কয়েনে Head আসার সম্ভাব্যতা কত?
* **Favorable Outcomes:** ছক্কায় জোড় সংখ্যাগুলো হলো ২, ৪, ৬। শর্ত অনুযায়ী এগুলোর সাথে Head যুক্ত হলে আমাদের কাঙ্ক্ষিত ফলাফলগুলো হবে: $\{2H, 4H, 6H\}$ (মোট ৩টি)।

**Calculation:**


$$P(\text{Even}, \text{Head}) = \frac{3}{12} = 0.25$$



# **Complement Probability (পূরক সম্ভাব্যতা)**


Complement Probability বা পূরক সম্ভাব্যতা মানে হলো— কোনো একটি নির্দিষ্ট ঘটনা **"না ঘটার"** সম্ভাব্যতা পরিমাপ করা।
যেহেতু আমরা জানি, কোনো ঘটনা ঘটার এবং না ঘটার মোট সম্ভাব্যতা সবসময় **$1$ (বা ১০০%)** হয়, তাই $1$ থেকে ঘটার সম্ভাব্যতা বাদ দিলেই না ঘটার সম্ভাব্যতা পাওয়া যায়।

**Mathematical Notation:**
কোনো ঘটনা যদি $A$ হয়, তবে তার না ঘটার ঘটনাকে সাধারণত $A'$, $A^c$, বা $\sim A$ দ্বারা প্রকাশ করা হয়।
**সূত্র:**

$$P(\sim A) = 1 - P(A)$$

#### **Example 1: Coin Flip (কয়েন টস)**

একটি কয়েন টস করলে Head ($H$) না আসার সম্ভাব্যতা কত?

* **ঘটার সম্ভাবনা:** $P(H) = \frac{1}{2} = 0.5$
* **না ঘটার সম্ভাবনা:**
$$P(\sim H) = 1 - P(H)$$


$$P(\sim H) = 1 - 0.5 = 0.5$$



*(বিকল্প লজিক: Head না আসা মানেই Tail আসা। আর Tail আসার সম্ভাবনাও $0.5$। আমরা বাকি ফলাফলগুলো যোগ করলেও একই উত্তর পাবো!)*

#### **Example 2: Die Roll (ছক্কা নিক্ষেপ)**

একটি ছক্কা নিক্ষেপ করলে বিজোড় (Odd) সংখ্যা না আসার সম্ভাব্যতা কত?

* Sample Space: $\{1, 2, 3, 4, 5, 6\}$
* **ঘটার সম্ভাবনা (Odd):** এখানে বিজোড় সংখ্যা ৩টি (১, ৩, ৫)। তাই $P(\text{Odd}) = \frac{3}{6} = 0.5$
* **না ঘটার সম্ভাবনা:**

$$P(\sim \text{Odd}) = 1 - P(\text{Odd})$$


$$P(\sim \text{Odd}) = 1 - 0.5 = 0.5$$

### **🔥 Machine Learning Aspect (লজিস্টিক রিগ্রেশনে এর ব্যবহার)**

ভাই, মনে আছে আমরা আগের নোটে **Logistic Regression** (Binary Classification) নিয়ে আলোচনা করেছিলাম?
সেখানে Complement Probability-এর কনসেপ্টটি সরাসরি ব্যবহার করা হয়!

ধরি, আপনার মডেল প্রেডিক্ট করছে একটি ছবি বিড়াল কি না।

* বিড়াল হওয়ার সম্ভাবনা $P(y=1)$ যদি হয় **$0.8$**
* তাহলে বিড়াল না হওয়ার (অর্থাৎ কুকুর হওয়ার) সম্ভাবনা মডেল আলাদা করে আর হিসাব করে না। মডেল শুধু **Complement Rule** ব্যবহার করে:

$$P(y=0) = 1 - P(y=1)$$


$$P(y=0) = 1 - 0.8 = 0.2$$



মেশিন লার্নিংয়ের বাইনারি ক্লাসিফিকেশনের মূল ভিত্তিই হলো এই Complement Rule!


# **Sum of Probability (সম্ভাব্যতার যোগসূত্র / Addition Rule)**

যখন আমাদের বের করতে হয় যে একাধিক ঘটনার মধ্যে যেকোনো একটি ঘটনা ঘটার সম্ভাব্যতা কত (যেমন: ঘটনা A **অথবা** B ঘটার সম্ভাবনা), তখন আমরা Sum of Probability বা যোগসূত্র ব্যবহার করি।

এটি মূলত দুটি বিষয়ের ওপর নির্ভর করে: ঘটনাগুলো কি Disjoint (বিচ্ছিন্ন) নাকি Joint (অবিচ্ছিন্ন)?

#### **1. Disjoint Events (পরস্পর বিচ্ছিন্ন ঘটনা)**

যখন দুটি ঘটনার মধ্যে কোনো সাধারণ বা কমন (Common) উপাদান থাকে না, তখন তাদের Disjoint বা Mutually Exclusive ঘটনা বলে।
**সূত্র:**

$$P(A \cup B) = P(A) + P(B)$$

* **Example 1:** একটি ছক্কা নিক্ষেপে ২ অথবা ৩ আসার সম্ভাব্যতা কত?
এখানে ২ এবং ৩ সম্পূর্ণ আলাদা ঘটনা (একসাথে ঘটা সম্ভব না)।

$$P(2 \text{ or } 3) = P(2) + P(3) = \frac{1}{6} + \frac{1}{6} = \frac{2}{6} = \frac{1}{3} \approx 0.33$$


* **Example 2:** ১ অথবা জোড় সংখ্যা (Even) আসার সম্ভাব্যতা কত?
জোড় সংখ্যাগুলো হলো {২, ৪, ৬}। এর মধ্যে ১ নেই, তাই এদের মধ্যে কোনো কমন উপাদান নেই (Disjoint)।

$$P(1 \text{ or Even}) = P(1) + P(\text{Even}) = \frac{1}{6} + \frac{3}{6} = \frac{4}{6} = \frac{2}{3} \approx 0.67$$



#### **2. Joint Events (অবিচ্ছিন্ন ঘটনা / Overlapping)**

যখন দুটি ঘটনার মধ্যে এক বা একাধিক কমন উপাদান থাকে, তখন শুধু যোগ করলে কমন উপাদানটি **দুইবার (Double Count)** যোগ হয়ে যায়। তাই মোট যোগফল থেকে কমন অংশটুকু ($P(A \cap B)$) একবার বিয়োগ করতে হয়।
**সূত্র (General Formula):**


$$P(A \cup B) = P(A) + P(B) - P(A \cap B)$$

* **Example 3:** একটি ছক্কা নিক্ষেপে ২ অথবা জোড় সংখ্যা (Even) আসার সম্ভাব্যতা কত?
* $A = $ ২ আসার সম্ভাবনা = $\frac{1}{6}$
* $B = $ জোড় সংখ্যা (২, ৪, ৬) আসার সম্ভাবনা = $\frac{3}{6}$
* **Double Count Issue:** জোড় সংখ্যার ভেতরে '২' আগে থেকেই আছে। তাই আমরা যদি সরাসরি $P(A) + P(B)$ করি, তবে '২' দুইবার যোগ হয়ে যাবে।
* **Intersection ($A \cap B$):** ২ এবং জোড় সংখ্যার মধ্যে কমন হলো শুধু '২'। এর সম্ভাবনা = $\frac{1}{6}$


**Calculation:**

$$P(2 \cup \text{Even}) = P(2) + P(\text{Even}) - P(2 \cap \text{Even})$$


$$P(2 \cup \text{Even}) = \frac{1}{6} + \frac{3}{6} - \frac{1}{6} = \frac{3}{6} = \frac{1}{2} = 0.5$$


# **Conditional Probability (শর্তযুক্ত সম্ভাব্যতা)**

**সংজ্ঞা:** Conditional Probability বলতে বোঝায়, একটি ঘটনা ঘটার সম্ভাব্যতা কতটুকু, যদি আমরা আগে থেকেই জানি যে **অন্য আরেকটি ঘটনা ইতিমধ্যে ঘটে গেছে (Given that)**।

**মূল কনসেপ্ট (Shrinking the Sample Space):**
যখনই কোনো শর্ত (Condition) জুড়ে দেওয়া হয়, তখন আমাদের মূল Sample Space থেকে শর্তের বাইরের ঘটনাগুলো বাদ (Discard) হয়ে যায় এবং Sample Space ছোট হয়ে আসে। নতুন এই ছোট Sample Space-এর ওপর ভিত্তি করেই আমরা সম্ভাব্যতা বের করি।

#### **Base Example: Tossing 2 Coins (২টি কয়েন নিক্ষেপ)**

কোনো শর্ত ছাড়া পরপর ২টি কয়েন টস করলে আমাদের ফলাফলগুলো হয়:

* **Total Sample Space ($S$):** $\{HH, HT, TH, TT\}$
* **পরপর ২টি Head আসার সাধারণ সম্ভাব্যতা:**
$$P(H,H) = \frac{1}{4} = 0.25$$



#### **Scenario 1: The Condition is Met (শর্ত যখন পূরণ হয়)**

**প্রশ্ন:** যদি আমরা জানি যে **প্রথম কয়েনটি Head (H)** এসেছে, তবে **দ্বিতীয়টি Tail (T)** আসার সম্ভাব্যতা কত?

* **Condition:** প্রথম কয়েনটি $H$ এসেছে।
* **Shrinking the Space:** যেহেতু প্রথমটি $H$ এসেছে, তাই যেসব ফলাফলে প্রথমে $T$ আছে ($TH, TT$) সেগুলো সম্পূর্ণ বাদ হয়ে যাবে।
* **New Sample Space:** $\{HH, HT\}$ (মোট ফলাফল এখন ২টি)
* **Favorable Outcome:** দ্বিতীয়টি $T$ আছে এমন ফলাফল হলো $\{HT\}$ (মাত্র ১টি)।

**Calculation:**


$$P(\text{2nd is } T \mid \text{1st is } H) = \frac{1}{2} = 0.5$$


*(নোট: যদি প্রশ্ন হতো "প্রথমটি হেড হলে দুটোই হেড ($H,H$) আসার সম্ভাবনা কত?", তাহলেও উত্তর $1/2$ বা $0.5$ হতো, কারণ নতুন স্পেসে $HH$ আছে ১ বার।)*

#### **Scenario 2: The Impossible Event (অসম্ভব ঘটনা)**

**প্রশ্ন:** যদি **প্রথমটি Tail (T)** আসে, তবে **পরপর দুটি Head (H,H)** আসার সম্ভাব্যতা কত?

* **Condition:** প্রথম কয়েনটি $T$ এসেছে।
* **New Sample Space:** $\{TH, TT\}$ (কারণ $H$ দিয়ে শুরু হওয়া ফলাফলগুলো বাদ)।
* **Favorable Outcome:** এই নতুন স্পেসের মধ্যে $HH$ আসার কোনো সুযোগই নেই, তাই অনুকূল ফলাফল $0$।

**Calculation:**


$$P(H,H \mid \text{1st is } T) = \frac{0}{2} = 0$$

---

**Summary:** Conditional Probability-তে আমরা সবসময় ডানদিকের বা শর্তের ঘটনাটিকে (Given event) ভিত্তি ধরে আমাদের জগত বা Sample Space-কে ছোট করে নিই, এবং তারপর সাধারণ নিয়মে হিসাব করি।

# **Derivation of Conditional Probability Formula (সূত্র কীভাবে এল?)**

Screenshot from 2026-05-20 08-24-28.png


আপনার চার্ট অনুযায়ী আমাদের কাছে ডেটা আছে:

* মোট ছাত্র-ছাত্রী বা Sample Space, $n(S) = 100$
* মোট ছাত্র, $n(\text{Male}) = 50$
* এমন ছাত্র যারা পাইথন পছন্দ করে (Male AND Python), $n(\text{Male} \cap \text{Python}) = 30$

#### **Step 1: The Logical Approach (আপনার বের করা লজিক)**

আপনি হিসাব করে বের করেছেন যে, একজন ছাত্রের পাইথন পছন্দ করার সম্ভাব্যতা:


$$P(\text{Python} \mid \text{Male}) = \frac{\text{ছাত্রদের মধ্যে পাইথন পছন্দ করে (30)}}{\text{মোট ছাত্র (50)}}$$


একে আমরা ম্যাথের ভাষায় বা সংখ্যায় (n) লিখতে পারি:


$$P(\text{Python} \mid \text{Male}) = \frac{n(\text{Male} \cap \text{Python})}{n(\text{Male})}$$


#### **Step 2: The Magic Trick (Number থেকে Probability-তে রূপান্তর)**

স্ট্যাটিস্টিকসের নিয়ম অনুযায়ী, কোনো কিছুর Number বা Count-কে যদি মোট Sample Space ($100$) দিয়ে ভাগ করা হয়, তবে সেটি Probability-তে পরিণত হয়।
সূত্র বানানোর জন্য আমরা এখন সমীকরণের ওপর এবং নিচ—উভয় পাশকেই মোট ছাত্রছাত্রী $n(S)$ দিয়ে ভাগ করব:

$$= \frac{\frac{n(\text{Male} \cap \text{Python})}{n(S)}}{\frac{n(\text{Male})}{n(S)}}$$

$$= \frac{\frac{30}{100}}{\frac{50}{100}}$$

#### **Step 3: Translating back to Probability (সূত্রে রূপান্তর)**

এবার একটু খেয়াল করে দেখুন তো, এই ভাগ করা অংশগুলো আসলে কী?

* উপরের অংশ $\frac{30}{100}$ মানে হলো: পুরো ১০০ জনের মধ্যে "ছাত্র এবং পাইথন লাভার" হওয়ার সম্ভাবনা। অর্থাৎ, এটি হলো **$P(\text{Male} \cap \text{Python})$**
* নিচের অংশ $\frac{50}{100}$ মানে হলো: পুরো ১০০ জনের মধ্যে "ছাত্র" হওয়ার সম্ভাবনা। অর্থাৎ, এটি হলো **$P(\text{Male})$**

তাহলে উপরের সমীকরণে এই মানগুলো বসালে আমরা পাচ্ছি:


$$P(\text{Python} \mid \text{Male}) = \frac{P(\text{Male} \cap \text{Python})}{P(\text{Male})}$$

#### **4. The Generalized Formula (চূড়ান্ত সূত্র)**

এখান থেকেই Conditional Probability-এর মূল বা সাধারণ (General) সূত্রটি তৈরি হয়েছে। যদি ঘটনা দুটি $A$ এবং $B$ হয়, এবং $B$ আগে থেকেই ঘটে থাকে (Given), তবে $A$ ঘটার সম্ভাবনা:

$$P(A \mid B) = \frac{P(A \cap B)}{P(B)}$$

> **সহজ কথায় মনে রাখার উপায়:** > যার সম্ভাব্যতা বের করতে হবে (যেমন $A \mid B$), তাদের দুজনের একসাথে ঘটার সম্ভাবনাকে ($A \cap B$) ওপরে বসাতে হবে। আর যে ঘটনাটি **আগে থেকেই ঘটে গেছে** ($B$), তার সম্ভাবনা দিয়ে নিচে ভাগ করতে হবে।
